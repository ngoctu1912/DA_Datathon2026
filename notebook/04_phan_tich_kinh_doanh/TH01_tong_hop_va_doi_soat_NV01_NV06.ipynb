```python
from pathlib import Path
from decimal import Decimal
from collections import Counter, defaultdict
from html.parser import HTMLParser
import csv, hashlib, os, re, sys, math, subprocess
import pandas as pd
import numpy as np
from IPython.display import display, Markdown
sys.dont_write_bytecode = True

import os

import sys

start = Path(os.environ.get("RETAIL_ANALYTICS_ROOT", Path.cwd())).resolve()
ROOT = next((path for path in (start, *start.parents)
             if (path / "notebook/ham_dung_chung.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Mở thư mục gốc của dự án hoặc đặt RETAIL_ANALYTICS_ROOT.")
sys.path.insert(0, str(ROOT / "notebook"))
from ham_dung_chung import read_notebook, stable_digest, notebook_prefix

from ham_dung_chung import read_notebook as load_notebook_for_audit, snapshot_files
from ham_dung_chung import repository_status
TARGET = ROOT/'notebook/04_phan_tich_kinh_doanh/TH01_tong_hop_va_doi_soat_NV01_NV06.ipynb'
THIS = load_notebook_for_audit(TARGET)
initial_notebook = THIS
HISTORICAL_BASELINE = THIS['metadata']['th011']['baseline_sha256']
def sha(p):
    with p.open('rb') as f: return hashlib.file_digest(f, 'sha256').hexdigest()
def snapshot():
    return {p.relative_to(ROOT).as_posix(): sha(p) for p in ROOT.rglob('*')
            if p.is_file() and '.git' not in p.relative_to(ROOT).parts
            and '__pycache__' not in p.parts and p != TARGET}
before = snapshot()
BASELINE = before.copy()
assert before == BASELINE, 'Tệp bảo vệ khác SHA-256 chụp trước nhiệm vụ; cần đối chiếu thay đổi.'
checks, table_log = [], []
def check(name, condition, evidence):
    ok = bool(condition)
    checks.append({'Kiểm chứng':name, 'Kết quả':'Đạt' if ok else 'Không đạt', 'Bằng chứng/phạm vi':evidence})
    assert ok, name
def show(rows, title):
    frame = rows if isinstance(rows, pd.DataFrame) else pd.DataFrame(rows)
    table_log.append(title)
    display(Markdown('**'+title+'**'))
    with pd.option_context('display.max_rows', None, 'display.max_columns', None, 'display.max_colwidth', None,
                           'display.width', 220):
        display(frame)
    return None
def fmt(v, digits=0):
    return f'{v:,.{digits}f}'.replace(',', '_').replace('.', ',').replace('_', '.')
def money(cents): return fmt(Decimal(int(cents))/100, 2)
def vi(s): return Decimal(str(s).strip().replace('.','').replace(',','.'))
NB = {}
for p in sorted((ROOT/'notebook').rglob('NV*.ipynb')):
    tag = p.name[:4]
    if tag in [f'NV{i:02}' for i in range(1,7)]:
        assert tag not in NB, 'Có nhiều notebook cùng mã; cần xác định nguồn.'
        doc = read_notebook(p)
        NB[tag] = {'path':p, 'doc':doc}
check('Đúng sáu notebook nguồn, đọc được dữ liệu notebook', set(NB)=={f'NV{i:02}' for i in range(1,7)}, '; '.join(sorted(NB)))
pd.set_option('display.max_colwidth', 160)
print('Gốc dự án:', ROOT)
print('Bảo vệ', len(BASELINE), 'tệp hiện có, gồm raw, NV01–NV06, profile_14_tables và .vscode/.')
```

### Phương pháp và truy vết bằng chứng

Số ô được đánh **từ 1** theo vị trí trong notebook hiện có; bảng bằng chứng lưu cả mã ô và SHA-256 nội dung. Đọc Markdown/code để xác định nghĩa và phạm vi; đọc bảng HTML đã lưu để lấy giá trị đối chiếu. Không chạy code trong notebook nguồn. Các số đề bài là **mốc tham khảo**, tách riêng khỏi số tính được.

K1: cùng định nghĩa, phạm vi và giá trị; K2: khác phạm vi; K3: khác định nghĩa; K4: thiếu bằng chứng; K5: sai lệch cùng định nghĩa đã được xác minh độc lập. K1 chỉ xác nhận phép tính trong dữ liệu; không xác nhận kế toán, tiền thực thu, SLA hoặc nhân quả. Với giá trị hiển thị đã làm tròn, dung sai là nửa đơn vị hiển thị và được ghi rõ.

```python
class Tables(HTMLParser):
    def __init__(self):
        super().__init__(); self.tables=[]; self.rows=[]; self.row=[]; self.buf=None
    def handle_starttag(self, tag, attrs):
        if tag=='table': self.rows=[]
        if tag=='tr': self.row=[]
        if tag in ('th','td'): self.buf=''
    def handle_data(self, data):
        if self.buf is not None: self.buf += data
    def handle_endtag(self, tag):
        if tag in ('th','td') and self.buf is not None:
            self.row.append(self.buf.strip()); self.buf=None
        if tag=='tr' and self.row: self.rows.append(self.row)
        if tag=='table' and self.rows: self.tables.append(self.rows)
saved = []
for nv, info in NB.items():
    for pos, cell in enumerate(info['doc']['cells'], 1):
        for out_no, out in enumerate(cell.get('outputs', []), 1):
            raw = out.get('data',{}).get('text/html')
            if raw:
                parser=Tables(); parser.feed(''.join(raw))
                for table_no, rows in enumerate(parser.tables, 1):
                    drop = rows[0][0]==''
                    saved.append({'NV':nv, 'Ô':pos, 'Mã ô':cell.get('id','Không có ID'),
                                  'Output':out_no, 'Bảng':table_no,
                                  'Cột':rows[0][1:] if drop else rows[0],
                                  'Dòng':[r[1:] if drop else r for r in rows[1:]]})
evidence = []
def row(nv, pos, label):
    hits = [(t,r) for t in saved if t['NV']==nv and t['Ô']==pos
            for r in t['Dòng'] if label in r[:3]]
    assert len(hits)==1, (nv,pos,label,len(hits))
    t,r=hits[0]
    key=f'{nv}/ô {pos}/output {t["Output"]}/bảng {t["Bảng"]}/{label}'
    evidence.append({'Mã bằng chứng':key, 'Đường dẫn':NB[nv]['path'].relative_to(ROOT).as_posix(),
                     'Ô (từ 1)':pos, 'ID ô':t['Mã ô'], 'Cột':' | '.join(t['Cột']),
                     'Dòng đã lưu':' | '.join(r),
                     'SHA-256 ô':stable_digest(NB[nv]['doc']['cells'][pos-1])})
    return r
def src(nv,pos): return ''.join(NB[nv]['doc']['cells'][pos-1]['source'])
def compact_execution(info):
    candidates=[]
    containers=[('metadata notebook',info['doc']['metadata'])]+[
        (f'metadata ô {i}',c.get('metadata',{})) for i,c in enumerate(info['doc']['cells'],1)]
    keep={'status','scope','two_fresh_kernels_completed','all_checks_passed',
          'all_applicable_checks_passed','checks_failed','verification_counts'}
    run_keep={'run','status','fresh_kernel','working_directory','cwd','code_cells_executed',
              'code_cells','elapsed_seconds','seconds'}
    for where,m in containers:
        for key,value in m.items():
            if not isinstance(value,dict) or not any(z in key.lower() for z in ['execution','run_all','nv06_']):
                continue
            nested=isinstance(value.get('execution'),dict)
            v=value['execution'] if nested else value
            small={k:x for k,x in v.items() if k in keep}
            if isinstance(v.get('runs'),list):
                small['runs']=[{k:x for k,x in run.items() if k in run_keep} for run in v['runs']]
            if small:
                candidates.append(where+'/'+key+('/execution' if nested else '')+': '+
                                  repr(small))
    return '\n'.join(candidates) or 'Chưa nhận diện metadata thực thi; xem output cuối và execution_count.'
```

**C:** nguồn chính phục vụ câu hỏi phân tích; **J:** nguồn JOIN/đối chiếu; **—:** không sử dụng trực tiếp. Nguồn JSON V01–V09 là bằng chứng kế thừa, không được coi là notebook đã đọc CSV tương ứng.
```python
roles = {
'NV01':{'C':['sales'],'J':[]},
'NV02':{'C':['customers','orders','order_items'],'J':['payments','geography']},
'NV03':{'C':['orders','order_items','payments','shipments'],'J':[]},
'NV04':{'C':['returns'],'J':['orders','order_items','payments','products','customers','geography']},
'NV05':{'C':['inventory'],'J':['products','orders','order_items','returns']},
'NV06':{'C':['promotions','order_items','customers','web_traffic'],'J':['orders','products','sales']}}
role_evidence={
'NV01':'Ô 4/6 đọc sales; ô 2 định nghĩa; ô 88/90 đối chiếu V02',
'NV02':'Ô 5 customers/orders; ô 47 items/payments; ô 65 geography',
'NV03':'Ô 3 đọc orders/items/payments; ô 26 shipments; returns chỉ tham chiếu V03',
'NV04':'Ô 2 đọc returns/orders/items/payments; .3 products; .4 customers/geography',
'NV05':'Ô 2 inventory/products/orders/items; ô 95 bổ sung returns',
'NV06':'Ô 2 đọc 7 bảng; ô 1 phân biệt trực tiếp và tham chiếu; .2–.4 sử dụng theo vai trò'}
paths={p.stem:p for p in sorted((ROOT/'data/raw').glob('*.csv'))}
matrix=[{'NV':nv,**{name:('C' if name in spec['C'] else 'J' if name in spec['J'] else '—') for name in paths},
         'Bằng chứng':role_evidence[nv]} for nv,spec in roles.items()]
show(matrix,'Ma trận NV01–NV06 × toàn bộ 14 CSV raw')
D={}; csv_audit=[]
for name,p in paths.items():
    with p.open(encoding='utf-8-sig',newline='') as f:
        reader=csv.reader(f); header=next(reader)
        assert all(header) and len(set(header))==len(header), name
        count=0
        for line in reader:
            assert len(line)==len(header), (name,count+2,'Sai độ rộng CSV')
            count+=1
    d=pd.read_csv(p,dtype='string',encoding='utf-8-sig',keep_default_na=False,
                  skip_blank_lines=False,on_bad_lines='error')
    check(name+': đọc đúng header/độ rộng/số dòng',list(d.columns)==header and len(d)==count,
          'Đối chiếu csv.reader độc lập với pandas; giữ nguyên mã và chuỗi rỗng.')
    D[name]=d; csv_audit.append({'CSV':name,'Dòng csv.reader':count,'Dòng pandas':len(d),'Số cột':len(header)})
# Chỉ phân tích trong bộ nhớ; giữ chuỗi nguồn D nguyên trạng.
keys={'customers':['customer_id'],'products':['product_id'],'orders':['order_id'],
      'order_items':['order_id','product_id'],'returns':['return_id'],
      'inventory':['product_id','snapshot_date'],'promotions':['promo_id'],'web_traffic':['date'],
      'geography':['zip'],'payments':['order_id'],'shipments':['order_id'],
      'reviews':['review_id'],'sales':['Date'],'sample_submission':['Date']}
grain={'customers':'Một hồ sơ khách','products':'Một mã sản phẩm','orders':'Một đơn',
       'order_items':'Một dòng hàng CSV; cặp đơn–sản phẩm không nhất thiết duy nhất',
       'returns':'Một bản ghi return_id; chưa xác nhận sự kiện vật lý/quyết toán',
       'inventory':'Một sản phẩm–ngày snapshot (tháng quan sát)',
       'promotions':'Một mô tả chương trình','web_traffic':'Một ngày có nhãn traffic_source',
       'geography':'Một ZIP có city/region/district','payments':'Một dòng payment, hiện một dòng/đơn',
       'shipments':'Một dòng shipment, hiện một dòng/đơn','reviews':'Một review_id',
       'sales':'Một ngày Revenue/COGS','sample_submission':'Một ngày mục tiêu mẫu, không phải giao dịch thực tế'}
date_specs={'customers':[('signup_date','%Y-%m-%d')],
            'orders':[('order_date','%Y-%m-%d')],'returns':[('return_date','%Y-%m-%d')],
            'inventory':[('snapshot_date','%Y-%m-%d')],'promotions':[('start_date','%Y-%m-%d'),('end_date','%Y-%m-%d')],
            'web_traffic':[('date','%Y-%m-%d')],'shipments':[('ship_date','%Y-%m-%d'),('delivery_date','%Y-%m-%d')],
            'reviews':[('review_date','%Y-%m-%d')],'sales':[('Date','%Y-%m-%d')],
            'sample_submission':[('Date','%Y-%m-%d')]}
dates={}; time_rows=[]; spans={}
for name,fields in date_specs.items():
    for col,date_format in fields:
        s=D[name][col]; missing=s.isin(['','NA','N/A','NULL','null','NaN','nan','None','<NA>'])
        parsed=pd.to_datetime(s.mask(missing),format=date_format,errors='coerce')
        invalid=int((~missing & parsed.isna()).sum())
        check(name+'.'+col+': ngày có giá trị đọc đúng định dạng',invalid==0,
              f'Định dạng {date_format}; {int(missing.sum())} trường thiếu vẫn giữ thiếu; không đổi thành 0.')
        dates[name,col]=parsed
        use=', '.join(nv for nv,r in roles.items() if name in r['C']+r['J']) or 'Không có NV01–NV06'
        span=f'{parsed.min().date()} → {parsed.max().date()}'
        spans.setdefault(name,[]).append(col+': '+span)
        time_rows.append({'Nguồn':name,'Trường ngày':col,'Từ':str(parsed.min().date()),'Đến':str(parsed.max().date()),
          'Ngày phân biệt':int(parsed.nunique()),'Thiếu':int(missing.sum()),'Không hợp lệ':invalid,'Nghiệp vụ':use})
limits_source={
'customers':'signup khác lịch sử orders; kênh là thuộc tính hồ sơ',
'products':'Giá/cogs hiện có, chưa xác nhận lịch sử và đơn vị tiền',
'orders':'Tất cả status; snapshot không có lịch sử; order_source không phải attribution đã xác minh',
'order_items':'Không có line_id; vị trí dòng chỉ dùng truy vết trong phiên, không khử trùng cặp',
'returns':'Return_id là bản ghi; refund chưa quyết toán; return_date khác tháng đặt',
'inventory':'Phạm vi sản phẩm/tháng không đầy đủ; snapshot khác luồng; không điền vắng',
'promotions':'Khoảng catalogue bao gồm hai biên; chưa xác nhận exposure/quy tắc tiền',
'web_traffic':'Một nhãn traffic_source/ngày, chưa phải bảng ngày×kênh; bounce_rate chưa xác minh thang',
'payments':'Không có ngày giao dịch/quyết toán; một dòng/đơn chỉ là thực trạng',
'shipments':'Ngày thiếu giữ thiếu; actual/planned và SLA chưa xác nhận',
'sales':'Revenue đối chiếu Gross, không phải Net; COGS chưa là lợi nhuận',
'geography':'ZIP là khóa tra cứu; không tự suy vùng khách khác vùng đơn',
'reviews':'Không sử dụng trực tiếp trong sáu NV; không mở phân tích review',
'sample_submission':'Không sử dụng trực tiếp; số mẫu tương lai không xác nhận Revenue/COGS thực tế'}
source_rows=[]
for name,d in D.items():
    k=keys[name]
    distinct=len(d[k].drop_duplicates())
    source_rows.append({'Nguồn CSV':'data/raw/'+name+'.csv','Định nghĩa dòng/độ hạt':grain[name],
      'Khóa kiểm tra':' + '.join(k),'Số dòng':len(d),'Khóa phân biệt':distinct,
      'Dòng khóa trùng dư':len(d)-distinct,'Dòng có khóa thiếu':int(d[k].eq('').any(axis=1).sum()),
      'Phạm vi thời gian':'; '.join(spans.get(name,[])) or 'Không có trường ngày trực tiếp',
      'Nghiệp vụ sử dụng':', '.join(nv for nv,r in roles.items() if name in r['C']+r['J']) or 'Không sử dụng',
      'Giới hạn':limits_source[name]})
show(source_rows,'Danh mục nguồn, độ hạt, khóa, phạm vi và giới hạn')
```

Số dòng payments/shipments hiện trùng với số đơn có mặt trong bảng đó không có nghĩa hai tập bằng toàn bộ orders.

```python
reference_counts={'customers':121930,'products':2412,'orders':646945,'order_items':714669,
                  'returns':39939,'inventory':60247,'promotions':50,'web_traffic':3652}
scale=[]
for name,d in D.items():
    k=keys[name]
    ref=reference_counts.get(name)
    scale.append({'Bảng':name,'Định nghĩa dòng':grain[name],'Số dòng tính từ CSV':len(d),
      'Khóa đếm phân biệt':' + '.join(k),'Số khóa phân biệt':len(d[k].drop_duplicates()),
      'Mốc đề bài (không tạo kết quả)':ref if ref is not None else 'Không có',
      'So mốc':'Khớp' if ref==len(d) else 'Chưa có mốc' if ref is None else 'Khác mốc — cần xem xét',
      'Phạm vi kiểm tra':'Toàn bộ dòng raw, không lọc/khử trùng',
      'Notebook dùng':', '.join(nv for nv,r in roles.items() if name in r['C']+r['J']) or 'Không sử dụng trực tiếp'})
show(scale,'Đối soát quy mô 14 CSV')
for name in D:
    if name!='order_items':
        check(name+': khóa ứng viên không trùng, không rỗng',
              not D[name].duplicated(keys[name]).any() and not D[name][keys[name]].eq('').any().any(),
              'Khóa được kiểm tra trên CSV hiện tại; không suy thành hợp đồng nghiệp vụ vĩnh viễn.')
print('Cặp order_id–product_id có dòng trùng dư:',int(D['order_items'].duplicated(keys['order_items']).sum()),
      '— vẫn giữ toàn bộ dòng.')
```

## TH01.1.4 — Đối soát tài chính

Gross = Σ(quantity × unit_price); Discount = Σ(discount_amount) **một lần/dòng**; Net = Gross − Discount, trên mọi trạng thái. Tiền dùng **ĐVT**, chưa xác nhận tên tiền tệ. Phép tính kiểm chứng dùng số nguyên phần trăm ĐVT sau khi kiểm tra Decimal chuyển chính xác; không làm tròn trung gian. Phép tính độc lập thứ hai đọc CSV bằng csv.DictReader và cộng Decimal.

Không gọi Net hay Payment là tiền thực thu. Không lấy Net trừ toàn bộ Refund để gán thành doanh thu kế toán. Chênh Revenue−COGS của NV01 là chênh số học trước giảm giá theo định nghĩa nguồn. Những JOIN dưới đây chỉ phục vụ đối soát và có kiểm soát lực lượng; không xuất ledger hoặc đặc trưng.

```python
def cents_exact(series, name):
    vals=series.map(Decimal)
    check(name+': tiền hữu hạn, chính xác đến phần trăm ĐVT',
          all(v.is_finite() and v*100==(v*100).to_integral_value() for v in vals),
          'Không điền NULL và không làm tròn khi chuyển số; dừng nếu trường không hợp lệ.')
    return vals.map(lambda v:int(v*100)).astype('int64')
def ints_exact(series,name):
    check(name+': số nguyên không thiếu',series.str.fullmatch(r'-?\d+').all(),'Giữ số nguồn; không thay giá trị thiếu.')
    return series.astype('int64')
orders,items,customers,ret=D['orders'],D['order_items'],D['customers'],D['returns']
q=ints_exact(items.quantity,'order_items.quantity')
price=cents_exact(items.unit_price,'order_items.unit_price')
disc=cents_exact(items.discount_amount,'order_items.discount_amount')
lines=pd.DataFrame({'order_id':items.order_id,'product_id':items.product_id,
                    'q':q,'gross':q*price,'discount':disc})
lines['net']=lines.gross-lines.discount
tot={k:int(lines[k].sum()) for k in ['gross','discount','net']}
payment_c=cents_exact(D['payments'].payment_value,'payments.payment_value')
refund_c=cents_exact(ret.refund_amount,'returns.refund_amount')
rq=ints_exact(ret.return_quantity,'returns.return_quantity')
item_order=lines.groupby('order_id',sort=False)[['gross','discount','net','q']].sum().reset_index()
pay_order=pd.DataFrame({'order_id':D['payments'].order_id,'payment':payment_c}).groupby('order_id',sort=False).payment.sum().reset_index()
ledger=orders[['order_id','customer_id','order_status','order_date']].merge(item_order,on='order_id',how='left',validate='one_to_one',indicator=True)
check('JOIN items đã tổng hợp không nhân bản đơn',len(ledger)==len(orders) and ledger.order_id.is_unique
      and ledger._merge.eq('both').all(),'orders LEFT JOIN item_order, validate=one_to_one; mọi đơn có items.')
ledger=ledger.drop(columns='_merge').merge(pay_order,on='order_id',how='left',validate='one_to_one',indicator=True)
check('JOIN payment không nhân bản hoặc mất đơn',len(ledger)==len(orders) and ledger._merge.eq('both').all(),
      'Payment được tổng hợp riêng, không nối dòng payment vào dòng items.')
check('Tổng Gross/Discount/Net được bảo toàn qua JOIN',all(int(ledger[k].sum())==v for k,v in tot.items()),'So với toàn bộ dòng items trước JOIN.')
check('Payment=Net ở từng đơn',ledger.payment.eq(ledger.net).all(),
      'Chỉ xác minh số học toàn bộ order_id; không xác nhận giao dịch được quyết toán.')
native_g=native_d=Decimal(0); native_q=0; native_line_count=0
native_by_status=Counter(); order_status=dict(zip(orders.order_id,orders.order_status))
with paths['order_items'].open(encoding='utf-8-sig',newline='') as f:
    for r in csv.DictReader(f):
        g=Decimal(r['quantity'])*Decimal(r['unit_price']); d=Decimal(r['discount_amount'])
        native_g+=g;native_d+=d;native_q+=int(r['quantity']);native_line_count+=1
        native_by_status[order_status[r['order_id']]]+=int((g-d)*100)
check('Gross/Discount/Net: phép cộng CSV Decimal độc lập',
      (native_g,native_d,native_g-native_d)==tuple(Decimal(tot[k])/100 for k in ['gross','discount','net']),
      'csv.DictReader/Decimal không dùng groupby hoặc tổng đã tính từ pandas.')
check('Số dòng và quantity: đối chứng độc lập',native_line_count==len(items) and native_q==int(q.sum()),'Không đổi dòng thành đơn hoặc quantity thành số bản ghi.')
sales_r=cents_exact(D['sales'].Revenue,'sales.Revenue')
sales_c=cents_exact(D['sales'].COGS,'sales.COGS')
daily_gross=ledger.groupby('order_date').gross.sum()
sales_dates=D['sales'].Date
check('sales.Revenue khớp Gross theo từng ngày đặt',set(daily_gross.index)==set(sales_dates)
      and np.array_equal(daily_gross.reindex(sales_dates).to_numpy(),sales_r.to_numpy()),
      'Đối chiếu đủ ngày; không chỉ dựa vào tổng toàn kỳ.')
money_refs={'gross':Decimal('16430476585.53'),'discount':Decimal('749607320.10'),'net':Decimal('15680869265.43')}
show([{'Chỉ tiêu':k,'Tính từ CSV (ĐVT)':money(v),'Mốc đề bài (ĐVT)':fmt(money_refs[k],2),
       'Chênh với mốc (ĐVT)':fmt(Decimal(v)/100-money_refs[k],2),'Phạm vi':'Mọi dòng items, mọi trạng thái'} for k,v in tot.items()],
     'Gross, Discount, Net — kết quả tính độc lập với mốc tham khảo')
fin_saved={}
r2=row('NV02',84,'Tổng tiền đơn ghi nhận')
for k in ['gross','discount','net']:
    m=re.search(k+r':\s*([\d.,]+)',r2[2]); assert m,(k,r2)
    fin_saved['NV02',k]=vi(m.group(1))
for k in ['gross','discount','net']:
    fin_saved['NV03',k]=vi(row('NV03',87,'C_'+k)[3])
    fin_saved['NV04',k]=vi(row('NV04',105,k)[3])
for k,code in [('gross','M22'),('discount','M23'),('net','M24')]:
    fin_saved['NV06',k]=vi(row('NV06',118,code)[3])
rev_saved=vi(row('NV01',90,'Tổng Revenue toàn kỳ')[1])
cross=[]
def rec(metric,a,b,va,vb,unit,grain,scope_a,scope_b,kind,why,proof,next_step,refs):
    cross.append({'Mã đối soát':f'DS{len(cross)+1:03}','Chỉ tiêu':metric,'Notebook A':a,'Notebook B':b,
      'Giá trị A':va,'Giá trị B':vb,'Đơn vị':unit,'Cấp quan sát':grain,
      'Phạm vi và mẫu số A':scope_a,'Phạm vi và mẫu số B':scope_b,'Kết quả đối soát':kind,
      'Giải thích khác biệt':why,'Mức độ xác minh':proof,'Hướng xử lý tiếp':next_step,'Bằng chứng':refs})
for nv in ['NV02','NV03','NV04','NV06']:
    for k in ['gross','discount','net']:
        check(nv+': '+k+' khớp phép tính raw',fin_saved[nv,k]==Decimal(tot[k])/100,
              'Output đã lưu, tiền hai chữ số; so chính xác với csv/Decimal.')
for nv in ['NV03','NV04','NV06']:
    for k in ['gross','discount','net']:
        rec(k,'NV02',nv,fmt(fin_saved['NV02',k],2),fmt(fin_saved[nv,k],2),'ĐVT','Dòng → đơn → toàn kỳ',
            f'{len(items)} dòng, {len(orders)} đơn, mọi trạng thái','Cùng tập dòng/đơn, không lọc status','K1',
            'Cùng công thức, tổng tiền và phạm vi; Discount không nhân quantity.',
            'Output hai notebook + CSV Decimal và JOIN kiểm soát','Giữ định nghĩa khi đưa sang TH01.2',
            f'NV02 ô 48/84; {nv} ô '+{'NV03':'87','NV04':'26/105','NV06':'2/32/118'}[nv]+'; raw order_items.csv')
check('NV01 Revenue khớp Gross raw',rev_saved==Decimal(tot['gross'])/100,'NV01 ô 2/90 và đối chiếu từng ngày sales–items.')
rec('Revenue / Gross','NV01','NV02',fmt(rev_saved,2),fmt(fin_saved['NV02','gross'],2),'ĐVT','Ngày / dòng hàng tổng toàn kỳ',
    f'{len(D["sales"])} ngày sales','Mọi trạng thái, toàn kỳ items','K1',
    'Revenue trong NV01 là Gross theo V02; đã kiểm tra lại theo ngày.','Output + CSV từng ngày','Giữ nhãn Gross/Revenue kèm định nghĩa',
    'NV01 ô 2/90; NV02 ô 48/84; sales.csv + order_items.csv + orders.csv')
rec('Revenue so Net','NV01','NV06',fmt(rev_saved,2),fmt(fin_saved['NV06','net'],2),'ĐVT','Tổng toàn kỳ',
    'Gross trước giảm giá','Sau giảm giá, chưa trừ Refund','K3','Chênh chính là Discount; hai đại lượng khác định nghĩa.',
    'Công thức + CSV Decimal','Không thay Revenue bằng Net khi tổng hợp','NV01 ô 2/90; NV06 ô 118 M22–M24')
rec('Payment so Net','NV03','NV04',money(int(ledger.payment.sum())),money(tot['net']),'ĐVT','Đơn → toàn kỳ',
    'Payment ghi nhận, mọi order_id','Net số học, mọi order_id','K3','Bằng nhau từng đơn trong nguồn hiện tại; chưa xác nhận thực thu.',
    'Output + CSV từng đơn','Cần chính sách quyết toán ở TH01.2','NV03 ô 87 C_payment/C_exact; NV04 ô 105 payment/net')
formula_review=[
{'NV':'NV01','Định nghĩa/cấp':'sales ngày, Revenue=Gross theo V02; không có Discount dòng trực tiếp','Code/Markdown':'Ô 2, 6, 88, 90','Kiểm tra':'Revenue từng ngày khớp Gross raw; COGS không đổi sang lợi nhuận'},
{'NV':'NV02','Định nghĩa/cấp':'gross=quantity*unit_price; discount=discount_amount; net=gross-discount','Code/Markdown':'Ô 47–48, 56–57, 84','Kiểm tra':'Tổng hợp items/payments theo đơn trước one_to_one; phạm vi A mọi status, B delivered'},
{'NV':'NV03','Định nghĩa/cấp':'Gross/Discount/Net và Payment ở cấp đơn','Code/Markdown':'Ô 2–3, 14, 47, 67, 87','Kiểm tra':'Discount cả dòng; không nhân quantity; shipment tổng hợp riêng; không dùng returned làm returns'},
{'NV':'NV04','Định nghĩa/cấp':'GrossOrder/DiscountOrder/NetOrder; Refund tổng riêng theo đơn/cặp','Code/Markdown':'Ô 25–26, 105','Kiểm tra':'Tổng riêng ba bảng rồi one_to_one; không trừ Refund thành thực thu'},
{'NV':'NV06','Định nghĩa/cấp':'gross_c=qty*cents(unit_price); discount_c=cents(discount_amount)','Code/Markdown':'Ô 2, 10, 32, 118','Kiểm tra':'Bảng lượt không chứa tiền dòng; tổng từ bảng dòng; không phân bổ tiền hai mã'}]
show(formula_review,'Rà soát công thức, đơn vị, JOIN và phạm vi tại nơi chỉ tiêu xuất hiện')
show([{'NV':nv,**{k:fmt(fin_saved[nv,k],2) for k in ['gross','discount','net']},
      'Phạm vi':'Mọi trạng thái; ĐVT; chưa trừ Refund'} for nv in ['NV02','NV03','NV04','NV06']],
     'Tổng tiền đã lưu trong bốn notebook giao dịch')
```

## TH01.1.5 — Đơn hàng và trạng thái

Đếm distinct order_id, customer_id và số dòng riêng. Delivered là nhãn snapshot; “có ngày delivery_date ghi nhận” là một tập khác và chưa xác nhận giao hàng thực tế. Không suy lịch sử chuyển trạng thái từ nhãn hiện tại. Đơn có mã dùng hợp order_id của dòng có ít nhất một vị trí mã, không cộng số đơn distinct từng chương trình.

```python
promo_present=items[['promo_id','promo_id_2']].ne('')  # Ô mã rỗng chỉ nhận diện không tham chiếu, không điền tiền.
nslots=promo_present.sum(axis=1)
check('Mã không rỗng không có ký hiệu NULL chữ',not items[['promo_id','promo_id_2']].isin(['NA','NULL','null','NaN','nan']).any().any(),
      'Phải xem lại quy tắc có mã nếu nguồn chứa token thiếu khác chuỗi rỗng.')
sets={
'Mọi đơn':set(orders.order_id),
'Status delivered':set(orders.loc[orders.order_status.eq('delivered'),'order_id']),
'Status cancelled':set(orders.loc[orders.order_status.eq('cancelled'),'order_id']),
'Status returned':set(orders.loc[orders.order_status.eq('returned'),'order_id']),
'Có returns':set(ret.order_id),
'Có mã trên dòng':set(items.loc[nslots.gt(0),'order_id']),
'Có payment_value':set(D['payments'].loc[D['payments'].payment_value.ne(''),'order_id']),
'Có dòng shipment':set(D['shipments'].order_id),
'Có delivery_date ghi nhận':set(D['shipments'].loc[dates['shipments','delivery_date'].notna(),'order_id'])}
check('Khóa ngoại items/returns/payment/shipment về orders',
      all(set(D[name].order_id)<=sets['Mọi đơn'] for name in ['order_items','returns','payments','shipments']),
      'So tập khóa, không đánh đồng độ phủ giữa các bảng.')
check('Khóa khách của orders có trong customers',set(orders.customer_id)<=set(customers.customer_id),'JOIN many_to_one, customers.customer_id duy nhất.')
status=orders.order_status.value_counts(dropna=False)
status_money=ledger.groupby('order_status')[['gross','discount','net']].sum()
status_rows=[]
for s,count in status.items():
    sv=row('NV03',87,'AC_'+s)
    saved_count=int(vi(re.match(r'([\d.]+)',sv[3]).group(1)))
    check('Trạng thái '+s+': count và Net độc lập',saved_count==count and
          int(status_money.at[s,'net'])==native_by_status[s], 'NV03 ô 87 + Counter từ CSV + groupby cấp đơn.')
    status_rows.append({'order_status':s,'Đơn':int(count),'% tổng đơn':fmt(Decimal(int(count))*100/len(orders),4),
      'Gross (ĐVT)':money(status_money.at[s,'gross']),'Discount (ĐVT)':money(status_money.at[s,'discount']),
      'Net (ĐVT)':money(status_money.at[s,'net']),'Nguồn':'NV03 ô 87 AC_'+s+'; orders và items raw'})
    rec('Số đơn '+s,'NV03','CSV độc lập',fmt(saved_count),fmt(count),'Đơn distinct','order_id',
        'Mọi đơn có nhãn snapshot '+s,'Cùng phạm vi, không lịch sử trạng thái','K1','Cùng nhãn, mẫu và số đếm.',
        'Output + CSV/Counter','Giữ nghĩa snapshot','NV03 ô 87 AC_'+s+'; orders.csv')
show(status_rows,'Đơn và tiền theo mọi trạng thái snapshot')
n_buyers=orders.customer_id.nunique()
r2_buy=row('NV02',84,'Có/chưa có đơn')
r3_buy=row('NV03',87,'A01')
r2_scale=row('NV02',84,'Khách và đơn')
total_saved={
    'NV02':int(vi(re.search(r'Đơn:\s*([\d.]+)',r2_scale[2]).group(1))),
    'NV03':int(vi(re.match(r'([\d.]+)',r3_buy[3]).group(1))),
    'NV04':int(vi(row('NV04',105,'orders')[3])),
    'NV06':sum(int(vi(part)) for part in row('NV06',118,'M62')[3].split('/'))}
check('Tổng đơn trong NV02/NV03/NV04/NV06 khớp CSV',
      all(v==len(orders) for v in total_saved.values()),
      'Trích output từng notebook; NV06 cộng hai tập trong/ngoài giao web đã phân hoạch; mọi trạng thái.')
for nv in ['NV03','NV04','NV06']:
    rec('Tổng số đơn','NV02',nv,fmt(total_saved['NV02']),fmt(total_saved[nv]),'Đơn distinct','order_id',
        'Toàn orders, mọi trạng thái','Cùng toàn kỳ; NV06 lấy trong + ngoài giao web' if nv=='NV06' else 'Cùng toàn orders, mọi trạng thái',
        'K1','Cùng định nghĩa và phạm vi; số đếm kiểm chứng trực tiếp CSV.',
        'Output hai notebook + distinct order_id từ CSV','Giữ tổng và mẫu số khi tổng hợp',
        'NV02 ô 84 Khách và đơn; '+{'NV03':'NV03 ô 87 A01','NV04':'NV04 ô 105 orders','NV06':'NV06 ô 118 M62 (trong/ngoài web)'}[nv])
all_customers2=int(vi(re.search(r'Khách:\s*([\d.]+)',r2_scale[2]).group(1)))
all_customers6=int(vi(row('NV06',118,'M41')[3]))
check('Tổng khách NV02/NV06 khớp CSV',all_customers2==all_customers6==len(customers),
      'Toàn bộ customer_id; không dùng tập khách có đơn làm mẫu số này.')
rec('Tổng hồ sơ khách','NV02','NV06',fmt(all_customers2),fmt(all_customers6),'Khách','customer_id',
    'Toàn customers, gồm khách chưa có đơn','Cùng snapshot customers','K1',
    'Cùng khóa và nguồn hồ sơ; khác chỉ tiêu khách có đơn.','Output + distinct customer_id từ CSV',
    'Giữ riêng khách toàn bộ và khách có đơn','NV02 ô 84 Khách và đơn; NV06 ô 118 M41')
saved_buy2=int(vi(re.search(r'Có đơn:\s*([\d.]+)',r2_buy[2]).group(1)))
saved_buy3=int(vi(re.search(r'đơn;\s*([\d.]+)',r3_buy[3]).group(1)))
check('Khách có đơn khớp NV02/NV03/raw',saved_buy2==saved_buy3==n_buyers,'distinct customer_id từ mọi orders.')
rec('Khách có ít nhất một đơn','NV02','NV03',fmt(saved_buy2),fmt(saved_buy3),'Khách','customer_id',
    'Mọi trạng thái, khách có ≥1 đơn','Cùng phạm vi, không gồm khách không đơn','K1',
    'Cùng mẫu, distinct customer_id.','Output + CSV độc lập','Không gọi là conversion','NV02 ô 84; NV03 ô 87 A01')
show([{'Tập đơn':name,'Distinct order_id':len(v),'Phạm vi/định nghĩa':name} for name,v in sets.items()],
     'Các tập đơn được giữ riêng')
set_comparisons=[]
for a,b in [('Status returned','Có returns'),('Status delivered','Có delivery_date ghi nhận'),
            ('Có mã trên dòng','Có payment_value'),('Status delivered','Có dòng shipment')]:
    set_comparisons.append({'Tập A':a,'Tập B':b,'A':len(sets[a]),'B':len(sets[b]),
      'Giao A∩B':len(sets[a]&sets[b]),'Chỉ A':len(sets[a]-sets[b]),'Chỉ B':len(sets[b]-sets[a]),
      'Kết luận':'Không tự coi tương đương; ý nghĩa và độ phủ khác nhau'})
show(set_comparisons,'Đối chiếu tập hợp order_id — không suy tương đương từ tổng số')
delivered_net=int(ledger.loc[ledger.order_status.eq('delivered'),'net'].sum())
saved_deliv_net=vi(re.search(r'Net B:\s*([\d.,]+)',row('NV02',84,'Net và AOV delivered')[2]).group(1))
check('NV02 delivered Net khớp raw',saved_deliv_net==Decimal(delivered_net)/100,'Cùng nhãn delivered; không xác nhận thực thu.')
rec('Net mọi status / delivered','NV03','NV02',money(tot['net']),money(delivered_net),'ĐVT','Tổng tiền cấp đơn',
    f'{len(orders)} đơn, mọi trạng thái',f'{len(sets["Status delivered"])} đơn nhãn delivered','K2',
    'NV02 phạm vi B chỉ delivered; hai tổng không bắt buộc bằng nhau.',
    'Output + lọc đúng nhãn từ CSV','Ghi chính sách lọc A/B ở TH01.2','NV03 ô 87 C_net; NV02 ô 84 Net và AOV delivered')
```

## TH01.1.6 — Hoàn trả và hoàn tiền

NV03 dùng nhãn returned và kết quả tham chiếu V03; NV04 dùng bản ghi returns. NV05 dùng returns theo sản phẩm–tháng trả. Các cấp dòng, đơn và cặp đơn–sản phẩm được đối soát riêng. Refund so Net/Payment theo **cùng order_id** hoặc cùng cặp đã tổng hợp, không JOIN bảng returns thô vào items thô rồi cộng tiền.

Giá trị Refund ghi nhận chưa xác minh đã hoàn tiền; return_quantity chưa xác minh nhập lại kho. Các quan sát Refund>Payment cần làm rõ định nghĩa và quy trình, không tự gán là sai số dữ liệu.

```python
ret_work=pd.DataFrame({'return_id':ret.return_id,'order_id':ret.order_id,'product_id':ret.product_id,
                       'refund':refund_c,'return_quantity':rq,'return_date':dates['returns','return_date']})
ret_order=ret_work.groupby('order_id',sort=False)[['refund','return_quantity']].sum().reset_index()
ret_ledger=ledger.drop(columns='_merge').merge(ret_order,on='order_id',how='inner',validate='one_to_one')
check('JOIN Refund cấp đơn không nhân tiền',len(ret_ledger)==ret.order_id.nunique()
      and int(ret_ledger.refund.sum())==int(refund_c.sum()),'INNER JOIN chỉ tập có returns; đơn không returns không bị điền tiền=0.')
item_pair=lines.groupby(['order_id','product_id'],sort=False)[['gross','discount','net','q']].sum().reset_index()
ret_pair=ret_work.groupby(['order_id','product_id'],sort=False)[['refund','return_quantity']].sum().reset_index()
pair_ledger=ret_pair.merge(item_pair,on=['order_id','product_id'],how='left',validate='one_to_one',indicator=True)
check('Khóa cặp returns có trong items, không nhân Refund',pair_ledger._merge.eq('both').all()
      and len(pair_ledger)==len(ret_pair) and int(pair_ledger.refund.sum())==int(refund_c.sum()),
      'Tổng hợp cả hai phía theo order_id+product_id trước JOIN; giữ nguyên dòng nguồn.')
native_refund=Decimal(0);native_rq=0;native_rids=set();native_ros=set()
with paths['returns'].open(encoding='utf-8-sig',newline='') as f:
    for r in csv.DictReader(f):
        native_refund+=Decimal(r['refund_amount']);native_rq+=int(r['return_quantity'])
        native_rids.add(r['return_id']);native_ros.add(r['order_id'])
check('Refund/quantity/đơn returns: CSV độc lập',native_refund==Decimal(int(refund_c.sum()))/100
      and native_rq==int(rq.sum()) and native_ros==sets['Có returns'] and len(native_rids)==len(ret),
      'csv.DictReader + Decimal + tập khóa; không dựa vào JOIN.')
return_metrics={'return_rows':len(ret),'has':ret.order_id.nunique(),'pairs':len(ret_pair),
                'quantity':int(rq.sum()),'refund':Decimal(int(refund_c.sum()))/100,
                'missing_returned':len(sets['Status returned']-sets['Có returns']),
                'over':int(ret_ledger.refund.gt(ret_ledger.payment).sum())}
return_rows=[]
for k,value in return_metrics.items():
    saved_value=vi(row('NV04',105,k)[3])
    check('NV04 '+k+' khớp CSV',Decimal(int(value))==saved_value if isinstance(value,(int,np.integer)) else value==saved_value,
          'NV04 ô 105, đúng cấp đếm và cùng tập nguồn.')
    return_rows.append({'Chỉ tiêu':k,'NV04 đã lưu':fmt(saved_value,2 if k=='refund' else 0),
      'Tính trực tiếp CSV':fmt(value,2 if k=='refund' else 0),
      'Đơn vị':'ĐVT' if k=='refund' else 'Lượng nguồn' if k=='quantity' else 'Cặp đơn–sản phẩm' if k=='pairs' else 'Đơn' if k in ['has','missing_returned','over'] else 'Bản ghi',
      'Bằng chứng':'NV04 ô 105/'+k+'; returns/orders/items/payments CSV'})
show(return_rows,'Quy mô và Refund — đối chiếu NV04 với CSV')
rec('returned / đơn có returns','NV03','NV04',fmt(len(sets['Status returned'])),fmt(len(sets['Có returns'])),'Đơn','order_id',
    'Nhãn order_status=returned','Có ≥1 bản ghi returns, mọi trạng thái','K3',
    f'Chỉ status returned: {len(sets["Status returned"]-sets["Có returns"])}; chỉ returns: {len(sets["Có returns"]-sets["Status returned"])}.',
    'Tập order_id độc lập + output','Thống nhất nghĩa trạng thái và sự kiện ở TH01.2',
    'NV03 ô 87 AC_returned; NV04 ô 105 has/missing_returned')
nv5_rq=vi(row('NV05',121,'M57')[3])
check('NV04/NV05 cùng tổng return_quantity',nv5_rq==return_metrics['quantity'],'Cùng returns.csv, không gọi là hàng nhập kho.')
rec('Tổng return_quantity','NV04','NV05',fmt(return_metrics['quantity']),fmt(nv5_rq),'Lượng nguồn','Dòng returns',
    'Toàn returns, chưa xác minh đơn vị vật lý giữa sản phẩm','Cùng returns; tổng theo tháng trả ở phần đối chiếu kho','K1',
    'Cùng trường và phạm vi; chưa xác nhận hoàn nhập kho.','Output + CSV độc lập','Cần cầu nối nhập kho',
    'NV04 ô 105 quantity; NV05 ô 121 M57')
show([{'Phạm vi':'Đơn có returns','Đơn':len(ret_ledger),'Refund (ĐVT)':money(int(ret_ledger.refund.sum())),
       'Net (ĐVT)':money(int(ret_ledger.net.sum())),'Payment (ĐVT)':money(int(ret_ledger.payment.sum())),
       'Refund/Payment (%)':fmt(Decimal(int(ret_ledger.refund.sum()))*100/int(ret_ledger.payment.sum()),4),
       'Refund>Payment (đơn)':int(ret_ledger.refund.gt(ret_ledger.payment).sum())},
      {'Phạm vi':'Cặp đơn–sản phẩm có returns','Đơn':ret_ledger.order_id.nunique(),
       'Refund (ĐVT)':money(int(pair_ledger.refund.sum())),'Net (ĐVT)':money(int(pair_ledger.net.sum())),
       'Payment (ĐVT)':'Không phân bổ Payment đơn cho sản phẩm',
       'Refund/Payment (%)':'Không áp dụng','Refund>Payment (đơn)':'Không áp dụng'}],
     'Refund và giá trị mua theo cùng cấp — không phân bổ Payment thiếu căn cứ')
rec('Refund / tiền hoàn đã quyết toán','NV04','Nguồn cần bổ sung',money(int(refund_c.sum())),'Chưa xác minh','ĐVT','Bản ghi / giao dịch quyết toán',
    'refund_amount trong returns','Thiếu ngày/chứng từ quyết toán','K4','Chưa có nguồn độc lập chứng minh dòng tiền đã chuyển.',
    'Chỉ xác minh số học của refund_amount','Yêu cầu định nghĩa và sổ đối soát ở TH01.2','NV04 ô 25/105; schema returns/payments')
```

## TH01.1.7 — Inventory và bán hàng

Kế thừa NV05, đối chiếu tại cấp **product_id–tháng**. Quantity nhóm theo **tháng đặt**, returns nhóm theo **tháng trả**, inventory theo **tháng snapshot**. Không điền cặp vắng bằng 0. Tỷ lệ khớp có mẫu số là phần giao có quan sát ở cả hai phía, không phải toàn bộ products hay inventory.

Residual = (stock_t − stock_trước) − (received_t − sold_t), chỉ tính khi cùng mã và cách đúng một tháng. Đây là giả thuyết số học: snapshot cuối tháng và luồng đúng kỳ chưa được xác nhận; residual bằng 0 chưa xác nhận đối soát sổ kho thực tế. Pearson cao cũng không chứng minh hai trường cùng nghĩa.

```python
tx=lines[['order_id','product_id','q']].merge(orders[['order_id','order_date']],on='order_id',how='left',validate='many_to_one')
check('JOIN dòng items với orders không nhân dòng/lượng',len(tx)==len(items) and int(tx.q.sum())==int(q.sum()),
      'orders.order_id duy nhất; many_to_one; toàn bộ dòng và trạng thái được giữ.')
tx['month']=pd.to_datetime(tx.order_date,format='%Y-%m-%d').dt.to_period('M').astype(str)
txpm=tx.groupby(['product_id','month'],sort=False).q.sum().reset_index()
inv=D['inventory'][['product_id','snapshot_date']].copy()
inv['month']=dates['inventory','snapshot_date'].dt.to_period('M').astype(str)
for field in ['stock_on_hand','units_received','units_sold']:
    inv[field]=ints_exact(D['inventory'][field],'inventory.'+field)
check('Inventory một dòng mỗi sản phẩm–tháng',not inv.duplicated(['product_id','month']).any(),'Kiểm tra lại độ hạt tháng thay vì mặc định từ khóa snapshot_date.')
ik=['product_id','month']
ijoin=inv.merge(txpm,on=ik,how='outer',validate='one_to_one',indicator=True)
common=ijoin.loc[ijoin._merge.eq('both')].copy()
check('JOIN inventory/quantity bảo toàn tập và giá trị thiếu',
      len(ijoin)==len(set(map(tuple,inv[ik].to_numpy()))|set(map(tuple,txpm[ik].to_numpy())))
      and ijoin.loc[ijoin._merge.eq('left_only'),'q'].isna().all()
      and ijoin.loc[ijoin._merge.eq('right_only'),'units_sold'].isna().all(),
      'Outer one_to_one; không dùng fillna(0); cặp vắng giữ NA.')
ordered=inv.sort_values(['product_id','month']).copy()
ordered['month_no']=pd.PeriodIndex(ordered.month,freq='M').asi8
previous=ordered.groupby('product_id',sort=False)
ordered['gap']=ordered.month_no-previous.month_no.shift()
ordered['prev_stock']=previous.stock_on_hand.shift()
adjacent=ordered.loc[ordered.gap.eq(1)].copy()
adjacent['residual']=(adjacent.stock_on_hand-adjacent.prev_stock.astype('int64'))-(adjacent.units_received-adjacent.units_sold)
inv_metrics={'M01':len(inv),'M03':inv.product_id.nunique(),'M04':len(D['products']),
             'M06':inv.month.nunique(),'M41':len(inv),'M42':len(txpm),'M43':len(common),
             'M44':int(ijoin._merge.eq('left_only').sum()),'M45':int(ijoin._merge.eq('right_only').sum()),
             'M46':int(common.units_sold.eq(common.q).sum()),
             'M47':float(common.units_sold.eq(common.q).mean()*100),
             'M48':float(common.units_sold.corr(common.q)),
             'M53':len(adjacent),'M54':int(adjacent.residual.eq(0).sum()),'M55':int(ordered.gap.gt(1).sum())}
inv_rows=[]
for k,value in inv_metrics.items():
    r=row('NV05',121,k); old=vi(r[3]); tol=Decimal('0.0000005') if k in ['M47','M48'] else Decimal(0)
    check('NV05 '+k+' khớp CSV',abs(Decimal(str(value))-old)<=tol,
          'NV05 ô 121; dung sai 0,0000005 chỉ cho tỷ lệ/hệ số hiển thị 6 chữ số.')
    inv_rows.append({'Mã NV05':k,'Chỉ tiêu':r[2],'Đã lưu':r[3],
                     'Tính lại':fmt(value,6 if k in ['M47','M48'] else 0),
                     'Mẫu số/phạm vi':'Cặp giao' if k in ['M43','M46','M47','M48'] else
                      'Cùng mã, đúng liền tháng' if k in ['M53','M54'] else 'Cặp có mặt trong nguồn tương ứng',
                     'Nguồn':'NV05 ô 121/'+k+'; raw inventory/orders/order_items'})
# Đối chứng khác cách tính: csv/dict cho quantity theo tháng và vòng lặp snapshot.
native_order_month=dict(zip(orders.order_id,orders.order_date.str[:7]))
native_pm=defaultdict(int)
with paths['order_items'].open(encoding='utf-8-sig',newline='') as f:
    for r in csv.DictReader(f):
        native_pm[r['product_id'],native_order_month[r['order_id']]]+=int(r['quantity'])
check('Quantity sản phẩm–tháng: dict CSV độc lập',native_pm==dict(zip(map(tuple,txpm[ik].to_numpy()),txpm.q)),
      'Không dùng lại groupby hoặc bảng nguồn của NV05.')
native_snap=defaultdict(list)
with paths['inventory'].open(encoding='utf-8-sig',newline='') as f:
    for r in csv.DictReader(f):
        y,m=map(int,r['snapshot_date'][:7].split('-'))
        native_snap[r['product_id']].append((y*12+m,int(r['stock_on_hand']),int(r['units_received']),int(r['units_sold'])))
native_res=[]
for product,obs in native_snap.items():
    obs.sort()
    for a,b in zip(obs,obs[1:]):
        if b[0]-a[0]==1:native_res.append((b[1]-a[1])-(b[2]-b[3]))
check('Residual đúng cặp liền tháng: vòng lặp CSV độc lập',len(native_res)==len(adjacent)
      and Counter(native_res)==Counter(adjacent.residual),'Không bắc cầu qua tháng vắng; không thêm returns vào received.')
show(inv_rows,'Đối soát trọng yếu inventory với bán hàng — giữ cấp sản phẩm–tháng')
rec('units_sold / order_items.quantity','NV05','NV01–NV03 (nguồn giao dịch)',
    fmt(int(common.units_sold.sum())),fmt(int(common.q.sum())),'Lượng nguồn','Sản phẩm–tháng',
    f'{len(common)} cặp giao, inventory.units_sold','Cùng cặp giao, quantity mọi status theo tháng đặt','K3',
    f'Khớp chính xác {inv_metrics["M46"]}/{len(common)} cặp ({fmt(inv_metrics["M47"],4)}%); Pearson {fmt(inv_metrics["M48"],6)} không chứng minh tương đương.',
    'CSV/dict độc lập; nghĩa nghiệp vụ chưa xác nhận','Xác minh kỳ luồng và chính sách xuất kho ở TH01.2',
    'NV05 ô 95/97/99/121 M43–M48; raw inventory/orders/items')
rec('Độ phủ sản phẩm','NV05','Nguồn products',fmt(inv.product_id.nunique()),fmt(len(D['products'])),'Sản phẩm','product_id',
    'Có ≥1 snapshot inventory','Toàn bộ products','K2','Inventory chỉ phủ một phần products.',
    'CSV tập khóa + NV05 output','Không suy tồn 0 cho sản phẩm vắng','NV05 ô 121 M03–M05; inventory/products.csv')
rec('Residual snapshot / cân đối kho thực tế','NV05','Nguồn cần bổ sung',
    f'{inv_metrics["M54"]}/{len(adjacent)} cặp residual=0','Chưa xác nhận','Cặp','Hai snapshot cùng mã liền tháng',
    'Giả thuyết stock thay đổi = received−sold','Thiếu chứng từ nhập/xuất/điều chỉnh và nghĩa snapshot','K4',
    'Số học được tái lập nhưng chưa chứng minh tính đầy đủ của sổ kho.',
    'CSV vòng lặp độc lập cho số học; thiếu bằng chứng nghiệp vụ','Xác minh luồng, snapshot và hàng trả ở TH01.2',
    'NV05 ô 106–107/121 M53–M54')
```

## TH01.1.8 — Khuyến mãi và marketing

Bảng dòng hàng dùng để cộng tiền; bảng lượt mã dùng để đếm vị trí tham chiếu. Dòng hai mã có một khoản Discount; chưa đủ quy tắc để phân bổ cho từng chương trình. Đơn có mã gồm mọi dòng của đơn, khác tập dòng có mã.

Acquisition_channel là thuộc tính hồ sơ khách, traffic_source là nhãn ngày web, promo_channel là thuộc tính chương trình. Ngày nằm trong khoảng catalogue chưa là exposure. Discount chưa là chi phí marketing. Tương quan trước/sau trừ mean năm–tháng chỉ phục vụ đối soát mô tả NV06, không đánh giá tác động nhân quả và không tạo bảng đặc trưng.

```python
slot_refs=pd.concat([items.loc[promo_present[c],['order_id','product_id',c]].rename(columns={c:'promo_id'})
                     for c in ['promo_id','promo_id_2']],ignore_index=True)
check('Khóa mã tham chiếu tồn tại trong promotions',set(slot_refs.promo_id)<=set(D['promotions'].promo_id),
      'Từng vị trí mã; không dùng số dòng bảng dài để tính tiền.')
dual=nslots.eq(2)
native_slots=Counter();native_dual_discount=Decimal(0)
with paths['order_items'].open(encoding='utf-8-sig',newline='') as f:
    for r in csv.DictReader(f):
        slots=int(bool(r['promo_id']))+int(bool(r['promo_id_2']))
        native_slots[slots]+=1
        if slots==2:native_dual_discount+=Decimal(r['discount_amount'])
check('Lượt mã và Discount hai vị trí: CSV độc lập',
      sum(k*v for k,v in native_slots.items())==len(slot_refs)
      and native_slots[2]==int(dual.sum()) and native_dual_discount==Decimal(int(disc.loc[dual].sum()))/100,
      'Đếm vị trí khác dòng; Discount hai vị trí cộng một lần.')
marketing_counts={'M01':len(D['promotions']),'M11':len(items),'M12':int(nslots.eq(0).sum()),
                  'M13':int(nslots.eq(1).sum()),'M14':int(dual.sum()),'M15':int(nslots.gt(0).sum()),
                  'M16':len(slot_refs),'M17':len(sets['Có mã trên dòng']),'M41':len(customers)}
mark_rows=[]
for k,value in marketing_counts.items():
    r=row('NV06',118,k);old=vi(r[3])
    check('NV06 '+k+' khớp CSV',Decimal(value)==old,'Output NV06 ô 118, cùng đơn vị và phạm vi.')
    mark_rows.append({'Chỉ tiêu':r[2],'NV06 đã lưu':r[3],'Tính từ CSV':fmt(value),'Đơn vị':r[4],'Nguồn':'NV06 ô 118/'+k})
dual_saved=row('NV06',118,'M33')
dual_saved_money=vi(dual_saved[3].split(';')[0])
check('NV06 Discount hai mã khớp CSV',dual_saved_money==native_dual_discount,'Một lần mỗi dòng; không cộng tiền trên bảng lượt.')
mark_rows.append({'Chỉ tiêu':'Discount riêng dòng hai vị trí mã','NV06 đã lưu':fmt(dual_saved_money,2),
                  'Tính từ CSV':money(int(disc.loc[dual].sum())),'Đơn vị':'ĐVT','Nguồn':'NV06 ô 118/M33; order_items.csv'})
show(mark_rows,'Phân biệt chương trình, lượt mã, dòng, đơn và tiền')
show([{'Cách cộng Discount':'Đúng: mỗi dòng một lần','Discount (ĐVT)':money(tot['discount']),'Diễn giải':'Giá trị đối soát'},
      {'Cách cộng Discount':'Giả lập cộng một lần mỗi vị trí mã','Discount (ĐVT)':money(int((disc*nslots).sum())),
       'Diễn giải':'Chẩn đoán nguy cơ: cộng trùng tiền hai vị trí; không phải kết quả áp dụng'},
      {'Cách cộng Discount':'Giả lập nhân discount_amount với quantity','Discount (ĐVT)':money(int((disc*q).sum())),
       'Diễn giải':'Chẩn đoán sai công thức; nguồn định nghĩa giảm giá cả dòng'}],
     'Hai nguy cơ cộng sai được lượng hóa để kiểm tra — không thay đổi notebook nguồn')
check('Phần Discount cộng trùng theo vị trí bằng Discount tập hai mã',
      int((disc*nslots).sum())-tot['discount']==int(disc.loc[dual].sum())
      and int(disc.loc[nslots.eq(0)].sum())==0,'Chỉ là đồng nhất số học trong nguồn hiện tại; không gán quy tắc mọi giảm giá phải có mã.')
acq=customers.acquisition_channel.value_counts(dropna=False).sort_index()
saved_acq=row('NV06',118,'M43')[3]
acq_saved={part.split(':')[0].strip():int(vi(part.split(':')[1])) for part in saved_acq.split(';')}
check('NV06 cơ cấu acquisition khớp mọi nhãn',acq_saved==acq.to_dict(),'Hồ sơ khách, không phải traffic hoặc đơn attribution.')
acq_nv2=row('NV02',84,'Kênh đông khách và kênh có tỷ lệ có đơn cao nhất')
largest=re.search(r'Đông khách:\s*([^;]+)',acq_nv2[2]).group(1)
largest_n=int(vi(re.search(r'N kênh đông nhất:\s*([\d.]+)',acq_nv2[2]).group(1)))
check('NV02/NV06 kênh đông khách khớp',acq.idxmax()==largest and int(acq.max())==largest_n,'Cùng customers snapshot; NV02 ô 84 và NV06 ô 118 M43.')
rec('Khách thuộc kênh đông nhất','NV02','NV06',f'{largest}: {fmt(largest_n)}',f'{acq.idxmax()}: {fmt(acq.max())}',
    'Hồ sơ khách','customer_id','Toàn bộ customers theo acquisition_channel','Cùng nguồn hồ sơ','K1',
    'Cùng thuộc tính và mẫu số; chưa là hiệu quả thu hút khách.','Output + CSV value_counts',
    'Giữ nghĩa thuộc tính hồi cứu','NV02 ô 84; NV06 ô 118 M43; customers.csv')
show([{'acquisition_channel':label,'Khách':int(v),'Đơn vị':'Hồ sơ khách','Nguồn':'customers.csv; NV06 ô 118 M43'} for label,v in acq.items()],
     'Số hồ sơ khách theo acquisition_channel')
order_customer=orders[['order_id','customer_id','order_date']].merge(customers[['customer_id','signup_date']],on='customer_id',
                                                                 how='left',validate='many_to_one')
check('JOIN ngày signup không nhân đơn',len(order_customer)==len(orders) and order_customer.signup_date.notna().all(),
      'Khóa customer_id được kiểm tra duy nhất; không biến signup thành ngày mua đầu.')
order_customer['od']=pd.to_datetime(order_customer.order_date,format='%Y-%m-%d')
order_customer['sd']=pd.to_datetime(order_customer.signup_date,format='%Y-%m-%d')
pre_signup=int(order_customer.od.lt(order_customer.sd).sum())
pre_saved=row('NV06',118,'M47')[3]
check('Đơn trước signup khớp NV06',int(vi(re.search(r'Trước signup:\s*([\d.]+)',pre_saved).group(1)))==pre_signup,
      'So ngày đặt với ngày signup parsed đúng YYYY-MM-DD; không lọc bỏ đơn.')
web_days=set(dates['web_traffic','date'])
order_days=set(dates['orders','order_date'])
joint_days=web_days&order_days
inside=dates['orders','order_date'].isin(joint_days)
catalogue=set()
for start,end in zip(dates['promotions','start_date'],dates['promotions','end_date']):
    assert start<=end
    catalogue.update(pd.date_range(start,end,freq='D'))
n_catalogue=len(joint_days&catalogue)
daily_orders=orders.groupby('order_date').size()
web_audit=pd.DataFrame({'date':dates['web_traffic','date'],
                        'sessions':pd.to_numeric(D['web_traffic'].sessions,errors='raise')})
web_audit=web_audit.loc[web_audit.date.isin(joint_days)].copy()
web_audit['orders']=web_audit.date.dt.strftime('%Y-%m-%d').map(daily_orders)
check('Đối chiếu web–đơn đúng ngày có quan sát, không điền 0',web_audit.orders.notna().all()
      and len(web_audit)==len(joint_days) and web_audit.date.is_unique,'Giao ngày; không thêm lịch ngoài phạm vi.')
raw_p=float(web_audit.sessions.corr(web_audit.orders))
raw_s=float(web_audit.sessions.rank().corr(web_audit.orders.rank()))
month=web_audit.date.dt.to_period('M')
centered=web_audit[['sessions','orders']]-web_audit.groupby(month)[['sessions','orders']].transform('mean')
center_p=float(centered.sessions.corr(centered.orders))
center_s=float(centered.sessions.rank().corr(centered.orders.rank()))
import statistics
check('Pearson gốc đối chứng statistics độc lập',math.isclose(raw_p,statistics.correlation(web_audit.sessions,web_audit.orders),abs_tol=1e-12),
      'Cùng ngày/mẫu, một phép tính không dùng pandas.corr.')
web_saved=int(vi(row('NV06',118,'M50')[3].split(';')[0]))
in_saved=int(vi(row('NV06',118,'M62')[3].split('/')[0]))
cat_saved=int(vi(row('NV06',118,'M63')[3].split('/')[0]))
check('Web, đơn trong giao và ngày catalogue khớp NV06',
      web_saved==len(web_days) and in_saved==int(inside.sum()) and cat_saved==n_catalogue,
      'NV06 M50/M62/M63; ngày catalogue gồm cả hai biên, chỉ đếm trong giao web–đơn.')
for k,p,s in [('M69',raw_p,raw_s),('M70',center_p,center_s)]:
    r=row('NV06',118,k); text=r[3]
    old_p=vi(re.search(r'Pearson\s+(-?[\d.,]+)',text).group(1))
    old_s=vi(re.search(r'Spearman\s+(-?[\d.,]+)',text).group(1))
    check('NV06 '+k+': tương quan khớp cùng phép biến đổi',
          abs(Decimal(str(p))-old_p)<=Decimal('0.0000005') and abs(Decimal(str(s))-old_s)<=Decimal('0.0000005'),
          'Dung sai hiển thị 6 số; trừ mean theo năm–tháng, không theo tháng lịch gộp nhiều năm.')
show([{'Chỉ tiêu':'Đơn trước signup','Tính từ CSV':fmt(pre_signup),'Phạm vi':'Mọi đơn/khách, không lọc status','Nguồn':'NV06 ô 118 M47'},
      {'Chỉ tiêu':'Ngày web','Tính từ CSV':fmt(len(web_days)),'Phạm vi':'; '.join(spans['web_traffic']),'Nguồn':'NV06 M50'},
      {'Chỉ tiêu':'Ngày giao web–đơn','Tính từ CSV':fmt(len(joint_days)),'Phạm vi':'Ngày hiện diện ở cả hai nguồn','Nguồn':'NV06 ô 92/118'},
      {'Chỉ tiêu':'Đơn trong/ngoài giao web','Tính từ CSV':fmt(int(inside.sum()))+' / '+fmt(int((~inside).sum())),
       'Phạm vi':'order_id, không phải lượt web','Nguồn':'NV06 M62'},
      {'Chỉ tiêu':'Ngày catalogue trong giao web–đơn','Tính từ CSV':fmt(n_catalogue),'Phạm vi':'Hợp khoảng start–end, bao gồm hai biên','Nguồn':'NV06 M63'},
      {'Chỉ tiêu':'Pearson / Spearman sessions–đơn gốc','Tính từ CSV':fmt(raw_p,6)+' / '+fmt(raw_s,6),
       'Phạm vi':f'{len(joint_days)} ngày','Nguồn':'NV06 M69'},
      {'Chỉ tiêu':'Pearson / Spearman sau trừ mean năm–tháng','Tính từ CSV':fmt(center_p,6)+' / '+fmt(center_s,6),
       'Phạm vi':'Cùng ngày; giảm khác biệt mức theo tháng; chưa là kiểm soát mọi yếu tố','Nguồn':'NV06 M70'}],
     'Signup, phạm vi web, catalogue và tương quan trước/sau đối soát thời gian')
rec('Lượt mã / dòng có mã','NV06','NV06',fmt(len(slot_refs)),fmt(int(nslots.gt(0).sum())),'Lượt / dòng','Vị trí mã / dòng hàng',
    'Mỗi vị trí không rỗng là một lượt','Mỗi dòng có ít nhất một mã đếm một','K3',
    f'Chênh {int(dual.sum())} do dòng có hai vị trí mã; không phải sai lệch.',
    'CSV/Counter độc lập','Giữ ba cấp lượt/dòng/đơn riêng','NV06 ô 118 M14–M17')
rec('Đơn toàn kỳ / trong web','NV03','NV06',fmt(len(orders)),fmt(int(inside.sum())),'Đơn','order_id',
    'Toàn kỳ từ tháng 7/2012','Chỉ giao ngày có web từ 2013','K2',
    f'{int((~inside).sum())} đơn ngoài giao web, không được gán traffic=0.',
    'Output + tập ngày CSV','Ghi cửa sổ nguồn ở báo cáo','NV03 ô 87 A01; NV06 ô 118 M62')
rec('Acquisition / traffic / promo channel','NV02','NV06','Thuộc tính hồ sơ khách','Nhãn ngày web / thuộc tính chương trình',
    'Nhãn','Khách / ngày / chương trình','customers.acquisition_channel','web_traffic.traffic_source và promotions.promo_channel','K3',
    'Chuỗi nhãn có thể giống nhau nhưng thiếu cầu nối sự kiện để attribution.',
    'Schema và cách JOIN ở notebook nguồn','Xác minh định nghĩa kênh tại TH01.2','NV02 ô 72–73; NV06 ô 17–18/79–80/118')
```

## TH01.1.9 — Phạm vi thời gian

Đối chiếu min/max, số ngày có mặt và độ đầy đủ lịch riêng với độ đầy đủ vòng đời. Orders và sales bắt đầu 04/07/2012 nên năm 2012 không tương đương một năm đầy đủ. Web chỉ có từ 2013. Có ngày 31/12 trong nguồn không chứng minh mọi sự kiện sau đặt hàng đã được theo dõi đủ; ngày cuối orders không phải thời điểm trích xuất hệ thống.

NV02 mô tả thời gian quan sát theo khách và độ nhạy trạng thái, chưa tạo cohort mới trong TH01.1. Nếu chuyển sang so cohort, phải xác nhận thời điểm bắt đầu và thời gian theo dõi; không tự lấy signup_date làm ngày mua đầu, không coi các khoảng vắng là 0.

```python
show(time_rows,'Phạm vi ngày theo nguồn và nghiệp vụ — đọc trực tiếp từng trường')
show([{'NV':nv,
       'Nguồn chính':', '.join(spec['C']),
       'Nguồn JOIN/đối chiếu':', '.join(spec['J']) or 'Không có CSV phụ trực tiếp',
       'Phạm vi theo nguồn/trường ngày':'\n'.join(name+': '+('; '.join(spans[name]) if name in spans else
           'Không có ngày trực tiếp; chỉ quy kỳ khi JOIN đúng khóa với nguồn có ngày')
           for name in spec['C']+spec['J']),
       'Bằng chứng':role_evidence[nv]+'; min/max được tính từ từng CSV ở bảng trên'}
      for nv,spec in roles.items()],
     'Nguồn và phạm vi thời gian của từng notebook — không gán chung một đồng hồ')
od=dates['orders','order_date']; rd=dates['returns','return_date']
return_dates=ret[['return_id','order_id']].copy()
return_dates['rd']=rd
return_dates=return_dates.merge(orders[['order_id','order_date']],on='order_id',how='left',validate='many_to_one')
return_dates['od']=pd.to_datetime(return_dates.order_date,format='%Y-%m-%d')
cross_month=int(return_dates.rd.dt.to_period('M').ne(return_dates.od.dt.to_period('M')).sum())
cross_year=int(return_dates.rd.dt.year.ne(return_dates.od.dt.year).sum())
lag=(return_dates.rd-return_dates.od).dt.days
check('Ngày đặt–trả không nhân dòng; khác tháng/năm khớp NV04',
      len(return_dates)==len(ret) and cross_month==int(vi(row('NV04',105,'cross_month')[3]))
      and cross_year==int(vi(row('NV04',105,'cross_year')[3])),
      'Mỗi return_id nối đúng một ngày đặt; không gán return_date thành order_date.')
year_coverage=[]
for year in sorted(od.dt.year.unique()):
    days=set(od.loc[od.dt.year.eq(year)])
    full=set(pd.date_range(f'{year}-01-01',f'{year}-12-31'))
    year_coverage.append({'Năm':int(year),'Ngày orders có quan sát':len(days),'Ngày lịch cả năm':len(full),
                          'Đủ lịch cả năm':days==full,'Ngoài phạm vi năm quan sát':len(full-days)})
show(year_coverage,'Độ đầy đủ lịch orders — không ngoại suy năm biên')
time_warnings=[
('Năm 2012',f'{od.min().date()} là ngày đầu orders; {len(set(od.loc[od.dt.year.eq(2012)]))} ngày có quan sát',
 'NV01 giữ tăng trưởng chỉ giữa năm đủ lịch; không so 2012 như cả năm','NV01 ô 7–9; NV03 ô 87 A02'),
('Web so orders',f'{len(order_days-web_days)} ngày orders ngoài web; {int((~inside).sum())} đơn ngoài giao',
 'Chỉ đối chiếu phần giao; ngoài phạm vi giữ thiếu','NV06 ô 92/118 M50/M62'),
('Signup so ngày đặt',f'{pre_signup} đơn trước signup',
 'Chưa xác minh ý nghĩa signup; không xóa đơn hoặc dùng làm mốc cohort mua','NV02 ô 6; NV06 M47'),
('Ngày trả so ngày đặt',f'{cross_month} dòng khác tháng; {cross_year} dòng khác năm; độ trễ quan sát {lag.min()}–{lag.max()} ngày',
 'Chọn đúng trục thời gian; không cộng distinct đơn tháng trả để suy toàn kỳ','NV04 ô 105 cross_month/cross_year; NV05 ô 95'),
('Tháng cuối/follow-up',f'orders kết thúc {od.max().date()}; returns kết thúc {rd.max().date()}',
 'Có ngày cuối lịch chưa chứng minh đủ theo dõi hoàn trả; chưa biết thời điểm chốt/trích xuất','NV03 ô 86–87; NV04 .4/.5'),
('Khách/cohort',f'customers signup: {dates["customers","signup_date"].min().date()}–{dates["customers","signup_date"].max().date()}',
 'Thời gian quan sát không đều; không suy cohort đồng thời gian từ các tổng','NV02 ô 40/87/90'),
('Inventory snapshot',f'{inv.month.min()}–{inv.month.max()}; {int(ordered.gap.gt(1).sum())} khoảng qua tháng vắng',
 'Không nối phương trình một tháng qua khoảng vắng; không cộng snapshot thành luồng năm','NV05 ô 106–107/121'),
('Catalogue',f'{len(catalogue)} ngày trong hợp mọi khoảng; {n_catalogue} ngày thuộc giao web–đơn',
 'Bao gồm hai biên; lịch có catalogue chưa xác minh tiếp xúc','NV06 ô 92/118 M63')]
show([{'Nội dung':a,'Bằng chứng':b,'Giới hạn/điều kiện':c,'Nguồn notebook':d} for a,b,c,d in time_warnings],
     'Rủi ro so sánh thời gian và điều kiện giữ lại')
rec('Tháng ghi nhận returns / tháng đặt','NV04','NV05',
    f'{cross_month} dòng khác tháng đặt',f'{len(ret)} dòng theo tháng trả','Dòng returns','return_id; tháng',
    'NV04 phân biệt order_date và return_date','NV05 returns theo tháng return_date để đối chiếu snapshot','K3',
    'Hai trục thời gian có mục đích khác nhau; không tự chuyển returns về tháng đặt.',
    'Output + JOIN nhiều–một + ngày CSV','Ghi trục thời gian khi tổng hợp','NV04 ô 105 cross_month; NV05 ô 95')
```

## TH01.2.4 — Quy ước bán hàng và tài chính

Payment và Refund không có bằng chứng quyết toán chỉ vì khớp hoặc liên hệ với Net.

Trung bình giá trị đơn và trung vị giá trị đơn phải mang tên khác nhau; trung vị được tính trên các giá trị đơn đã tổng hợp, không lấy trung vị dòng hàng làm trung vị đơn. Tỷ số hai tổng khác trung bình các tỷ số.

```python
p012_fin=[
('KPI-001','Gross','Dòng gốc quantity × unit_price, cộng sau kiểm tra khóa','Không lấy sales.Revenue làm Net; sales.Revenue khớp Gross theo ngày','X1','TH01.1 DS010/DS011; NV01 ô 2'),
('KPI-002','Discount','discount_amount của cả dòng, cộng đúng một lần','Không nhân quantity, không cộng trên bảng lượt mã','X1','TH01.1 ô 12/20; NV06 ô 32'),
('KPI-003','Net số học','Gross−Discount cùng dòng/đơn/tập lọc','Không suy thực thu, lợi nhuận, dòng tiền thuần hoặc doanh thu kế toán','X1','TH01.1 DS003/006/009/011'),
('KPI-004','Payment ghi nhận','Tổng riêng payments theo order_id trước JOIN','Không có ngày giao dịch; cả cancelled có số ghi nhận; không xác nhận đã thu','X1','TH01.1 DS012; NV03 ô 87'),
('KPI-005','Refund ghi nhận','Tổng riêng returns theo order_id hoặc cặp trước so giá trị mua','Không tự gọi tiền đã trả lại, giá trị hàng nhập kho hoặc tính Net−Refund thành thực thu','X1','TH01.1 DS027; NV04 ô 25–26'),
('KPI-008/009/010/011','Tỷ số tiền','Tử/mẫu cùng tập đơn, cùng cửa sổ và mẫu số hợp lệ','Mẫu số 0/thiếu giữ không xác định; không gộp mẫu A/B','X2','TH01.1 DS024/036; NV02 ô 83'),
('KPI-011/012','Hai trọng số AOV','AOV gộp cho mỗi đơn trọng số 1; mean AOV khách cho mỗi khách trọng số 1','Mean/median trên đơn và trên khách phải ghi rõ; không thay thế nhau','X1','TH01.1 DS036; NV02 ô 84')]
p012_show([dict(zip(['KPI','Khái niệm','Quy ước','Giới hạn','Mức cho mục đích mô tả','Căn cứ'],r)) for r in p012_fin],
          'Định nghĩa tài chính và phạm vi sử dụng chung')
```

## TH01.2.6 — Khách hàng và ranh giới RFM

Hồ sơ khách, khách có đơn, khách có delivered và khách có đơn trong cửa sổ là những mẫu khác nhau. Với khách không đơn, tổng rỗng theo định nghĩa có thể báo riêng là “không có đơn trong nguồn”; không tự điền Recency, AOV hoặc trường tiền thiếu bằng 0. Tỷ lệ khách có đơn phải dùng toàn bộ hồ sơ làm mẫu số và không được gọi là conversion.

Trả khác tháng đặt không phải lỗi thời gian tự động.

Tỷ lệ đơn có returns không thay tỷ số lượng trả/lượng đặt. Các tỷ lệ phải nêu cửa sổ theo dõi; độ trễ quan sát lớn nhất không phải thời hạn trả chính thức. Không tự cộng return_quantity vào units_received.

## TH01.2.8 — Tồn kho và công thức ứng viên

Khóa nền là product_id + snapshot_date; khi quy về tháng phải kiểm tra tính duy nhất lần nữa. Không coi inventory là panel đầy đủ; tháng vắng giữ là không quan sát. Không cộng stock_on_hand nhiều tháng thành tổng tồn cả năm hoặc thay cho tồn bình quân.

Các quan hệ dưới được gọi là **quan hệ toán học thực nghiệm hoặc công thức ứng viên**, chưa phải công thức nghiệp vụ chính thức. Chúng phụ thuộc các trường thành phần, nên sự khớp không tạo bằng chứng độc lập về hiệu quả tồn kho.

```python
def p012_source(nv,pos,label):
    hits=[]
    for out in NB[nv]['doc']['cells'][pos-1].get('outputs',[]):
        html=out.get('data',{}).get('text/html')
        if not html:continue
        parser=Tables();parser.feed(''.join(html))
        for rows in parser.tables:
            drop=rows[0][0]==''
            head=rows[0][1:] if drop else rows[0]
            for rr in rows[1:]:
                values=rr[1:] if drop else rr
                if label in values[:3]:hits.append(dict(zip(head,values)))
    assert len(hits)==1,(nv,pos,label,len(hits))
    return hits[0]
p012_candidates=[
('KPI-045','fill_rate ≈ 1 − stockout_days/30','M38','5e-5','30 là mẫu số giả thuyết thực nghiệm, chưa xác nhận lịch/kho'),
('KPI-044','days_of_supply ≈ 30 × stock_on_hand/units_sold','M39','0,05 ngày','Chỉ xét mẫu số sold>0; chưa xác nhận nhu cầu/ngày hay tồn bình quân'),
('KPI-046','sell_through_rate ≈ units_sold/(stock_on_hand+units_sold)','M40','5e-5','Chỉ mẫu số dương; không xác nhận kỳ luồng hay vòng quay')]
p012_show([{'KPI':k,'Quan hệ ứng viên':f,'Dung sai đã dùng tại NV05':tol,
 'Bằng chứng số học đã lưu':str(p012_source('NV05',121,m)),
 'Mục đích và trạng thái':'X3: định nghĩa vận hành chưa chốt; quan hệ thực nghiệm chỉ có phạm vi dữ liệu NV05',
 'Điều kiện/giới hạn':why+'; không dùng chỉ báo phụ thuộc để xác minh vòng tròn'}
 for k,f,m,tol,why in p012_candidates],
 'Quan hệ toán học thực nghiệm — không phải công thức nghiệp vụ chính thức')
p012_show([
{'Nội dung':'Residual snapshot','Quy ước':'KPI-047, cùng mã và hai tháng cách đúng 1; bỏ phép chênh qua tháng vắng, không bỏ dòng nguồn',
 'Trạng thái theo mục đích':'X1 số học / X3 nghĩa snapshot; đối soát sổ kho thực tế vẫn K4',
 'Căn cứ':'TH01.1 DS030; NV05 M53–M54'},
{'Nội dung':'Units_sold và quantity','Quy ước':'Giữ riêng KPI-039 và KPI-015; mẫu số cặp giao phải được nêu',
 'Trạng thái theo mục đích':'X3 nghĩa units_sold; X4 nếu dùng thay quantity mà chưa có cầu nối',
 'Căn cứ':'TH01.1 DS028/V06'},
{'Nội dung':'Cờ và cực trị','Quy ước':'Giữ stockout/overstock/reorder và DOS nguyên trạng; không tự gắn nhãn vận hành',
 'Trạng thái theo mục đích':'X3 mô tả tín hiệu; KPI-043 X4 cho nhãn nhị phân',
 'Căn cứ':'NV05 M33/M35–M40/M56; V16'}],
 'Định nghĩa snapshot, lượng và cờ tồn kho theo mục đích')
```

## TH01.2.15 — Quyết định đối với đúng 16 vấn đề TH01.1

D1: có quy ước phân tích đủ căn cứ cho mục đích ghi rõ; D2: thống nhất tạm thời có điều kiện; D3: cần định nghĩa nghiệp vụ; D4: cần dữ liệu/kiểm chứng bổ sung; D5: không dùng cho mục tiêu hiện tại. D1 không đóng những ý nghĩa ngoài mục đích đã ghi; D2 không phải chính sách chính thức. Các yêu cầu metadata và dữ liệu bổ sung được tách riêng.

V09 chỉ dừng **phân bổ tiền hai mã theo chương trình**, vẫn cho phép báo tổng một lần trên dòng. V15 chỉ dừng **đánh giá tác động/hiệu quả nhân quả**, vẫn cho phép mô tả catalogue và tương quan.

## TH01.2.16 — Rà soát bốn K4 bằng kiểm tra có giới hạn

Thực hiện thêm các phép kiểm đọc trực tiếp CSV cho phần **có thể kiểm tra**: tổng Refund/schema, mẫu residual liền tháng, miền bounce và các nhãn status. Chúng không thay thế tài liệu định nghĩa hoặc chứng từ vận hành. Nếu phần nghiệp vụ không thể kiểm chứng với nguồn hiện có, giữ K4 và ghi rõ điều kiện; không sửa lịch sử DS027/030/037/038 trong TH01.1.

```python
with (ROOT/'data/raw/returns.csv').open(encoding='utf-8-sig',newline='') as stream:
    reader=csv.DictReader(stream);p012_return_fields=reader.fieldnames
    p012_refund_native=sum((Decimal(r['refund_amount']) for r in reader),Decimal(0))
with (ROOT/'data/raw/payments.csv').open(encoding='utf-8-sig',newline='') as stream:
    p012_payment_fields=next(csv.reader(stream))
p012_refund_saved=vi(p012_source('NV04',105,'refund')['Giá trị'])
p012_check('DS027: tổng Refund CSV/Decimal khớp output NV04',p012_refund_native==p012_refund_saved,
           'Chỉ kiểm số học; schema được ghi trong bảng mới, không chứng minh quyết toán.')
p012_targets=set(sorted(D['inventory'].product_id.unique())[:5]);p012_sample=defaultdict(list)
with (ROOT/'data/raw/inventory.csv').open(encoding='utf-8-sig',newline='') as stream:
    for r in csv.DictReader(stream):
        if r['product_id'] not in p012_targets:continue
        year,month=map(int,r['snapshot_date'][:7].split('-'))
        p012_sample[r['product_id']].append((year*12+month,int(r['stock_on_hand']),int(r['units_received']),int(r['units_sold'])))
p012_sample_res=[]
for product,observations in p012_sample.items():
    observations.sort()
    for prev,now in zip(observations,observations[1:]):
        if now[0]-prev[0]==1:p012_sample_res.append((now[1]-prev[1])-(now[2]-now[3]))
p012_check('DS030: mẫu CSV độc lập chỉ dùng cặp liền tháng',bool(p012_sample_res) and all(v==0 for v in p012_sample_res),
           f'{len(p012_sample_res)} cặp của 5 mã chọn theo khóa chuỗi; mẫu giới hạn, không nâng thành kiểm kê vật lý.')
with (ROOT/'data/raw/web_traffic.csv').open(encoding='utf-8-sig',newline='') as stream:
    p012_bounces=[Decimal(r['bounce_rate']) for r in csv.DictReader(stream)]
p012_bounce_saved=p012_source('NV06',118,'M59')['Giá trị'].split('/')
p012_check('DS037: miền bounce nguồn khớp min/max đã lưu',
           min(p012_bounces)==vi(p012_bounce_saved[0]) and max(p012_bounces)==vi(p012_bounce_saved[-1]),
           'Không dựa vào miền nhỏ hơn 1 để tự xác nhận tỷ lệ hay phần trăm.')
with (ROOT/'data/raw/orders.csv').open(encoding='utf-8-sig',newline='') as stream:
    p012_native_states={r['order_status'] for r in csv.DictReader(stream)}
p012_check('DS038: nhãn status đọc CSV độc lập khớp TH01.1',p012_native_states==set(status.index),
           'Tập nhãn không cung cấp định nghĩa đơn hợp lệ hoặc lịch sử chuyển trạng thái.')
p012_k4_evidence={
'DS027':('Đã kiểm tra tổng Refund bằng CSV/Decimal; returns có '+', '.join(p012_return_fields)+
         '; payments có '+', '.join(p012_payment_fields),
         'Chưa có chứng từ/trạng thái quyết toán và cấu phần tiền hoàn được xác nhận','Sổ giao dịch/chi trả, loại tiền, cấu phần và thời điểm quyết toán'),
'DS030':(f'Mẫu độc lập {len(p012_sample_res)} cặp đúng liền tháng residual=0; TH01.1 đã kiểm toàn cặp đủ điều kiện',
         'Không biết snapshot thực kiểm/cuối kỳ và tính đầy đủ luồng; không có chứng từ điều chỉnh',
         'Định nghĩa snapshot, received/sold, chứng từ kho, điều chỉnh và trả thực nhập'),
'DS037':(f'CSV cho miền bounce {min(p012_bounces)}–{max(p012_bounces)}, khớp output',
         'Miền số không chứng minh thang phân số/phần trăm hoặc mẫu số','Từ điển bounce, đơn vị và công thức theo phiên'),
'DS038':('Tập nhãn hiện tại: '+', '.join(sorted(p012_native_states)),
         'Snapshot không tự xác định tập đơn hợp lệ theo mục tiêu','Định nghĩa trạng thái, mục tiêu báo cáo, lịch sử sự kiện nếu cần')}
p012_k4_rows=[]
for r in p012_k4:
    new,why,need=p012_k4_evidence[r['Mã đối soát']]
    p012_k4_rows.append({'Mã gốc':r['Mã đối soát'],'Nguồn/chỉ tiêu A':r['Notebook A']+': '+r['Giá trị A'],
      'Nguồn/chỉ tiêu B':r['Notebook B']+': '+r['Giá trị B'],'Phạm vi A/B':r['Phạm vi và mẫu số A']+' / '+r['Phạm vi và mẫu số B'],
      'Lý do K4 gốc':r['Giải thích khác biệt'],'Bằng chứng bổ sung có thể kiểm tra':new,
      'Phân loại ban đầu':'K4','Kết luận mới':'Giữ cần xác minh; không thay đổi K4 lịch sử',
      'Độ tin cậy/phạm vi':'X1 cho phép kiểm số học/schema cụ thể; X3 cho nghĩa nghiệp vụ; X4 nếu suy quy trình chưa chứng minh',
      'Vì sao chưa chốt':why,'Cần bổ sung':need})
    p012_check(r['Mã đối soát']+': xác nhận nghiệp vụ thực tế',False,why+'; cần '+need,status='Không thể chạy')
p012_show(p012_k4_rows,'Rà soát bốn K4 — có kiểm tra bổ sung nhưng chưa đủ căn cứ đóng')
```

## TH01.2.18 — Kiểm tra từ điển với nguồn và kết quả đã lưu

Kiểm tra 15 chỉ tiêu trọng yếu bằng giá trị dẫn từ output nguồn và kết quả TH01.1 vừa thực thi. Công thức Gross/Discount/Net được đối chiếu thêm với cây cú pháp Python tại NV02 ô 48; không chỉ kiểm tra một chuỗi công thức tự viết trong từ điển.

Bảng ghi rõ cấp, bộ lọc, trục ngày, nguồn và giới hạn mức xác minh. Khớp số không nâng X3/X4 thành xác nhận vận hành. Nếu thấy mâu thuẫn phải để mở, không sửa nguồn. TH01.2 không tính lại toàn bộ nghiệp vụ hoặc dựng các phân phối mới.

```python
p012_fin_ast=ast.parse(src('NV02',48))
p012_assignments={}
for node in ast.walk(p012_fin_ast):
    if isinstance(node,ast.Assign):
        for target in node.targets:
            if isinstance(target,ast.Subscript) and isinstance(target.slice,ast.Constant):
                key=target.slice.value
                if key in ['gross','discount','net']:p012_assignments[key]=node.value
p012_g=p012_assignments['gross'];p012_di=p012_assignments['discount'];p012_ne=p012_assignments['net']
p012_check('Gross trong nguồn thật dùng quantity nhân unit_price',
 isinstance(p012_g,ast.BinOp) and isinstance(p012_g.op,ast.Mult)
 and isinstance(p012_g.left,ast.Attribute) and p012_g.left.attr=='quantity'
 and isinstance(p012_g.right,ast.Attribute) and p012_g.right.attr=='unit_price',
 'NV02 ô 48 AST; không sửa notebook nguồn.')
p012_check('Discount nguồn là giá trị cả dòng, không nhân quantity',
 isinstance(p012_di,ast.Attribute) and p012_di.attr=='discount_amount','NV02 ô 48 AST; đối chứng ví dụ hai mã ở bộ kiểm chứng.')
p012_check('Net nguồn là Gross trừ Discount',
 isinstance(p012_ne,ast.BinOp) and isinstance(p012_ne.op,ast.Sub)
 and p012_ne.left.attr=='gross' and p012_ne.right.attr=='discount','NV02 ô 48 AST.')
with (ROOT/'data/raw/web_traffic.csv').open(encoding='utf-8-sig',newline='') as stream:
    p012_sessions_native=sum(int(r['sessions']) for r in csv.DictReader(stream))
def p012_old_value(nv,pos,label,key='Giá trị'):
    return vi(p012_source(nv,pos,label)[key])
p012_critical=[
('KPI-001',Decimal(tot['gross'])/100,vi(p012_cross_by_id['DS001']['Giá trị A']),'TH01.1 DS001; NV02 ô 48'),
('KPI-002',Decimal(tot['discount'])/100,vi(p012_cross_by_id['DS002']['Giá trị A']),'TH01.1 DS002; NV06 ô 32'),
('KPI-003',Decimal(tot['net'])/100,vi(p012_cross_by_id['DS003']['Giá trị A']),'TH01.1 DS003; NV04 ô 26'),
('KPI-013',len(orders),vi(p012_cross_by_id['DS019']['Giá trị A']),'TH01.1 DS019'),
('KPI-014',len(items),p012_old_value('NV06',118,'M11'),'NV06 ô 118 M11'),
('KPI-020',n_buyers,vi(p012_cross_by_id['DS023']['Giá trị A']),'TH01.1 DS023'),
('KPI-028',len(ret_order),p012_old_value('NV04',105,'has'),'NV04 ô 105 has'),
('KPI-005',p012_refund_native,p012_refund_saved,'NV04 ô 105 refund; CSV/Decimal mới'),
('KPI-034',inv.product_id.nunique(),p012_old_value('NV05',121,'M03'),'NV05 ô 121 M03'),
('KPI-036',len(common),p012_old_value('NV05',121,'M43'),'NV05 ô 121 M43'),
('KPI-049',len(slot_refs),p012_old_value('NV06',118,'M16'),'NV06 ô 118 M16'),
('KPI-050',int(nslots.gt(0).sum()),p012_old_value('NV06',118,'M15'),'NV06 ô 118 M15'),
('KPI-051',len(sets['Có mã trên dòng']),p012_old_value('NV06',118,'M17'),'NV06 ô 118 M17'),
('KPI-059',p012_sessions_native,p012_old_value('NV06',118,'M51'),'NV06 ô 118 M51; CSV/int mới'),
('KPI-066',len(web_days),vi(p012_source('NV06',118,'M50')['Giá trị'].split(';')[0]),
 'Số ngày web: mẫu số KPI-066/067, không phải mean sessions; NV06 M50')]
p012_consistency=[]
for kid,value,old,source in p012_critical:
    k=p012_index[kid]
    p012_check(kid+': giá trị hoặc mẫu số dẫn đúng nguồn',Decimal(int(value))==old if isinstance(value,(int,np.integer)) else value==old,source)
    p012_consistency.append({'Mã':kid,'Đại lượng kiểm': 'Số ngày web (mẫu số)' if kid=='KPI-066' else k['Tên tiếng Việt'],
      'Tính/kiểm hiện tại':str(value),'Output đã lưu':str(old),'Công thức từ điển':k['Công thức'],
      'Cấp':k['Khóa/cấp quan sát'],'Đơn vị': 'Ngày' if kid=='KPI-066' else k['Đơn vị'],
      'Bộ lọc':k['Điều kiện lọc'],'Trục ngày':k['Trục thời gian'],
      'Mẫu số nếu có':k['Mẫu số'],'Mức không được nâng':k['Mức xác minh'],
      'Kết quả':'Khớp giá trị/mẫu số; phạm vi được nêu rõ','Nguồn':source})
p012_show(p012_consistency,'Kiểm tra 15 chỉ tiêu trọng yếu: giá trị, cấp, trạng thái, ngày và mẫu số')
```

## TH01.2.19 — Mức sẵn sàng cho TH01.3 và giai đoạn sau

“Sẵn sàng” chỉ nói về phép mô tả có tên/công thức/phạm vi đủ rõ. “Có điều kiện” yêu cầu giữ giới hạn; “chưa sẵn sàng” thiếu định nghĩa hoặc dữ liệu cho mục tiêu cụ thể. Những hàng Giai đoạn 2 dưới đây chỉ tiếp nhận điều kiện cần đáp ứng, chưa chọn ngày RFM, Monetary chính thức, nhãn, mô hình hoặc transformation.

**Cập nhật 15/11/2026:** đoạn trên và bảng bên dưới là trạng thái khi bàn giao TH01.2. # TH01.3 — Kết nối các phát hiện liên nghiệp vụ NV01–NV06
## TH01.3.1 — Kiểm tra nguồn và phương pháp tổng hợp

**Phạm vi duy nhất:** bổ sung TH01.3 sau 72 ô TH01.1–TH01.2. Các ghi chú phạm vi trong phần cũ là lịch sử bàn giao. Giữ nguyên JSON của 72 ô, metadata cấp notebook và mọi tệp khác. Không triển khai TH01.4 hoặc Giai đoạn 2.

Đọc lại kết luận cuối NV01 ô 97, NV02 ô 91, NV03 ô 96, NV04 ô 117, NV05 ô 131 và NV06 ô 130; chọn bằng chứng từ bảng tổng hợp đã lưu, đối chiếu với TH01.1 vừa chạy và dùng nguyên danh mục KPI/RULE của TH01.2. Số ô nguồn đánh từ **1**. Không chạy lại EDA của sáu notebook. Mỗi trích dẫn lưu đường dẫn, ID ô, vị trí output, dòng bằng chứng và SHA-256 JSON ô nguồn.

Phân biệt **A:** dữ liệu/số học theo quy ước; **B:** thống kê mô tả; **C:** giả thuyết cần kiểm tra; **D:** chưa đủ cơ sở kết luận. A/B/C/D này khác nhóm quy tắc hoặc giới hạn ở các notebook nguồn. **X1 không xác nhận quy trình ngoài dữ liệu.** Các bảng chủ đề là phép chiếu của bằng chứng và LINK, không tạo một từ điển chỉ tiêu khác.

```python
# Tên riêng TH01.3; không thay đổi biến, danh mục hoặc trạng thái lịch sử.
p013_doc=load_notebook_for_audit(TARGET)
p013_d0_kpi = p012_digest(p012_kpis)
p013_d0_rules = p012_digest(p012_rules)
p013_guard=p013_doc['cells'][72]['metadata']['th013_protection']
p013_checks=[];p013_tables=[];p013_provenance=[]
def p013_check(name,ok,evidence,state=None):
    state=state or ('Đạt' if bool(ok) else 'Không đạt')
    p013_checks.append({'Kiểm chứng TH01.3':name,'Trạng thái':state,'Bằng chứng/giới hạn':evidence})
    if state=='Không đạt':raise AssertionError(name)
def p013_show(rows,title):
    p013_tables.append(title)
    display(Markdown('**'+title+'**'))
    with pd.option_context('display.max_rows',None,'display.max_columns',None,'display.max_colwidth',None):
        display(pd.DataFrame(rows))
def p013_read(nv,pos,label):
    hits=[];cell=NB[nv]['doc']['cells'][pos-1]
    for out_no,out in enumerate(cell.get('outputs',[]),1):
        html=out.get('data',{}).get('text/html')
        if not html:continue
        parser=Tables();parser.feed(''.join(html))
        for table_no,table in enumerate(parser.tables,1):
            drop=table[0][0]=='';head=table[0][1:] if drop else table[0]
            for raw in table[1:]:
                values=raw[1:] if drop else raw
                if label in values[:3]:hits.append((dict(zip(head,values)),out_no,table_no))
    assert len(hits)==1,(nv,pos,label,len(hits))
    record,out_no,table_no=hits[0]
    locator=f'{nv}/ô {pos}/output {out_no}/bảng {table_no}/{label}'
    if locator not in {r['Mã bằng chứng'] for r in p013_provenance}:
        p013_provenance.append({'Mã bằng chứng':locator,'Notebook':NB[nv]['path'].relative_to(ROOT).as_posix(),
        'Ô nguồn':pos,'ID ô':cell['id'],'SHA-256 ô':p012_digest(cell),'Dòng bằng chứng':repr(record)})
    return record
p013_kpis=copy.deepcopy(p012_kpis);p013_rules=copy.deepcopy(p012_rules)
p013_kindex={r['Mã chỉ tiêu']:r for r in p013_kpis}
p013_rindex={r['Mã']:r for r in p013_rules}
p013_d0=p012_digest(p012_decision_rows);p013_k0=p012_digest(p012_cross)
p013_check('Đầu vào đúng 72 ô',p013_guard['cells']==72,'Ảnh chụp trước khi bổ sung, không đếm từ notebook đã mở rộng.')
p013_check('Bảo toàn dữ liệu notebook 72 ô',notebook_prefix(p013_doc['cells'][:72]) == notebook_prefix(initial_notebook['cells'][:72]),'Gồm nội dung nguồn và ID các ô.')
p013_check('Metadata cấp notebook giữ nguyên',stable_digest(p013_doc['metadata']) == stable_digest(initial_notebook['metadata']),'Audit mới chỉ đặt trong metadata ô TH01.3.')
p013_check('Đủ KPI/RULE kế thừa',len(p013_kpis)==68 and len(p013_rules)==30,'Không định nghĩa hoặc đổi trạng thái KPI mới.')
p013_check('38 đối soát và K1–K5 không đổi',len(p012_cross)==38 and p012_kcounts==Counter({'K1':23,'K2':3,'K3':8,'K4':4}),'K5=0; bảng TH01.1 ô 24.')
p013_check('16 quyết định D giữ nguyên',p012_dcounts==Counter({'D1':1,'D2':4,'D3':5,'D4':4,'D5':2}),'Quyết định TH01.2, không tự đóng vấn đề.')
p013_check('Phân bố X giữ nguyên',p012_xcounts==Counter({'X1':37,'X2':19,'X3':11,'X4':1}),'Theo mục đích chính, không áp X1 sang mục đích khác.')
p013_check('Bốn K4 đúng mã', {r['Mã đối soát'] for r in p012_k4}=={'DS027','DS030','DS037','DS038'},'Giữ nguyên phân loại gốc.')
# Đọc thêm bảng gốc NV06, phục vụ trích dẫn; không tính lại chiến dịch.
p013_m6={f'M{i:02}':p013_read('NV06',118,f'M{i:02}') for i in [4,11,14,15,16,17,22,23,24,25,33,37,38,39,40,41,42,43,44,45,47,50,51,52,53,58,59,62,63,64,65,66,69,70,72,73]}
p013_m5={f'M{i:02}':p013_read('NV05',121,f'M{i:02}') for i in [1,3,4,5,11,35,36,41,42,43,44,45,46,47,48,51,52,53,54,55,56,57]}
def p013_m(nv,key):return (p013_m6 if nv=='NV06' else p013_m5)[key]['Giá trị']
p013_origins=[
('NV01','sales ngày; đối chiếu Gross từ items','KPI-001; KPI-006; KPI-007','RULE-005; RULE-014; RULE-015','NV01 ô 90/92/97','Revenue là Gross; chênh COGS chưa là lợi nhuận'),
('NV02','customers và đơn theo customer_id','KPI-019; KPI-020; KPI-022; KPI-024; KPI-025','RULE-003; RULE-008; RULE-016; RULE-020','NV02 ô 84/86/91','Cơ hội quan sát khác nhau; chưa chọn RFM'),
('NV03','orders; items/payment tổng riêng; shipment','KPI-013; KPI-016; KPI-017; KPI-018','RULE-002; RULE-004; RULE-008; RULE-009','NV03 ô 87/89/96','Snapshot và ngày ghi nhận chưa xác minh vòng đời/SLA'),
('NV04','returns dòng/đơn/cặp và hai trục ngày','KPI-005; KPI-027; KPI-028; KPI-030; KPI-033','RULE-004; RULE-009; RULE-010; RULE-014','NV04 ô 105/107/117','Refund chưa là chi tiền; trả chưa là nhập kho'),
('NV05','inventory sản phẩm–tháng hiện diện','KPI-034; KPI-035; KPI-036; KPI-039; KPI-047','RULE-011; RULE-012; RULE-013; RULE-018; RULE-027','NV05 ô 121/123/131','KPI-039 X3 vận hành; residual X1 số học, K4 kho vẫn mở'),
('NV06','catalogue/lượt/dòng/đơn/khách/ngày web','KPI-049; KPI-050; KPI-051; KPI-055; KPI-059; KPI-068','RULE-006; RULE-021; RULE-023; RULE-024; RULE-029','NV06 ô 118/120/130','Không exposure, attribution hoặc hiệu quả nhân quả')]
p013_show([{'Nghiệp vụ':nv,'Phạm vi':scope,'Chỉ tiêu quan trọng':'; '.join(p013_kindex[k]['Tên tiếng Việt'] for k in ids.split('; ')),
 'Mã KPI liên quan':ids,'Quy tắc RULE liên quan':rules,'Phát hiện nguồn':source,'Mức độ xác minh':'; '.join(k+': '+p013_kindex[k]['Mức xác minh'] for k in ids.split('; ')),
 'Giới hạn sử dụng':limit} for nv,scope,ids,rules,source,limit in p013_origins], 'Ma trận nguồn – KPI/RULE – phạm vi sử dụng')
```

Vì vậy không đổi tên Revenue thành Net và không gọi Net là thực thu: chưa có xác nhận quyết toán, còn các nhãn cancelled/created/returned vẫn có giá trị ghi nhận.

## TH01.3.4 — Chủ đề 2: Khách hàng và hoạt động mua

Nhiều order_id cho thấy số đơn ghi nhận tích lũy lớn, chưa xác nhận nhiều lần mua độc lập hoặc lòng trung thành. Cửa sổ quan sát, số ngày đặt và trạng thái cần được xét cùng số đơn. Net khách bằng số đơn nhân AOV của chính khách theo cùng mẫu; tương quan số đơn–Net có thành phần cơ học. AOV gộp theo đơn khác mean AOV theo khách.

Signup_date là ngày hồ sơ chưa xác minh nguồn gốc; các đơn trước signup vẫn được giữ. Acquisition_channel chỉ phân nhóm hồi cứu giao dịch thuộc hồ sơ qua customer_id, không chứng minh nguồn tạo từng đơn. Khách có đơn / toàn hồ sơ là độ phủ giao dịch trong nguồn, thiếu chuỗi exposure–chuyển đổi để gọi conversion. Phân khúc có thể kế thừa định nghĩa cấp khách, nhưng còn phải chốt tập đơn, ngày chốt, cửa sổ và tiền; tại đây chưa chọn Monetary, Recency, Frequency chính thức hoặc triển khai RFM, phân cụm, retention.

## TH01.3.9 — Ma trận liên kết các phát hiện

**L1:** cùng chỉ tiêu đã đối soát; **L2:** bổ sung ngữ cảnh; **L3:** thống kê mô tả; **L4:** khác định nghĩa/phạm vi; **L5:** giả thuyết hoặc chưa đủ bằng chứng. Mỗi LINK nêu hai phát hiện, KPI/RULE, giới hạn và bước tiếp. L3 không là nhân quả; L5 chưa được chứng minh. Liên kết trong cùng NV phải thực sự bổ sung hai khía cạnh khác nhau cho lập luận liên nghiệp vụ.

## TH01.3.10 — Bức tranh nghiệp vụ tổng thể

Marketing và khách hàng mở đầu mạch kể nhưng chưa tạo được hành trình từ quảng cáo tới giao dịch. Customers có customer_id duy nhất và orders tham chiếu khóa này, cho phép mô tả đơn/Net thuộc hồ sơ khách. Web_traffic chỉ có số đo ngày; đối chiếu ngày sau tổng hợp không truy được phiên/người tạo đơn. Các nhãn kênh khách, web và catalogue giữ nghĩa riêng (LINK-003/015/016).

Đặt hàng nối với giá trị qua order_id: tổng hợp items về đơn trước khi ghép orders. Cặp đơn–sản phẩm là khóa tổng hợp thực nghiệm, chưa là ID dòng chính thức. Gross, Discount, Net dùng cùng tập dòng; sales.Revenue khớp Gross theo ngày. NV01 cung cấp góc thời gian, NV03/NV06 bổ sung status/mã, tạo cầu nối số học chưa xác nhận hiệu quả hoặc tiền đã thu (LINK-001/002/004).

Thanh toán và giao nhận có bản ghi gắn order_id; thực trạng một–một của payments/shipments không phải hợp đồng khóa vĩnh viễn. Payment=Net và khoảng ngày giao nhận mô tả được trên mẫu đủ cặp, nhưng thiếu ngày/trạng thái quyết toán, nghĩa actual/planned và lịch sử status. Không biến các bảng đã JOIN thành lịch sử xử lý hoàn chỉnh hoặc đo SLA chính xác (SRC-008/009/010; LINK-006/007).

Returns dùng return_id; tổng riêng về đơn hoặc đơn–sản phẩm trước đối chiếu để tránh nhân tiền. Nhãn returned khác hiện diện returns; return_date khác order_date về câu hỏi thời gian. Refund và return_quantity là tiền/lượng ghi nhận, thiếu bridge chi hoàn/thực nhập kho nên chưa tính thực thu hoặc tự tăng received (LINK-008/009; SRC-014).

Tồn kho khép lại mạch kể ở product_id–snapshot/tháng. Giao với quantity theo tháng đặt cho phép mô tả hai trường lượng; residual đúng liền tháng xác nhận nội bộ inventory. Tập mã thay đổi và tháng vắng giới hạn suy rộng. Residual0 không giải quyết lệch lượng bán hoặc xác nhận kho vật lý. Mạch này tổ chức câu chuyện bằng chứng, chưa là chuỗi sự kiện hay sơ đồ nhân quả được chứng minh (LINK-010–012/017/018).

```python
p013_flow=[
('Marketing → khách/đơn','customers; web_traffic; promotions; orders','customer_id nhiều–một; ngày sau tổng hợp 1–1','KPI-019/020/059/068','Đơn/Net theo hồ sơ; sessions–đơn theo ngày','Không bridge phiên/người/touchpoint → order','LINK-003/013/015/016'),
('Đặt hàng → giá trị','orders; order_items; sales','order_id; Date so order_date đã đối chiếu từng ngày','KPI-001/002/003/013/014/015','Tổng items về đơn; Revenue=Gross','Không line_id chính thức/quyết toán','LINK-001/002/004'),
('Giá trị → thanh toán/giao nhận','payments; shipments; orders','order_id thực trạng1–1, kiểm lại khi nguồn đổi','KPI-004/017/018','Payment=Net; khoảng ngày nguồn','Không ngày payment, lịch sử status hoặc nghĩa actual/SLA','SRC-008/009/010; LINK-007'),
('Đơn → hoàn trả','returns; orders; order_items','return_id; tổng riêng order_id/cặp trước JOIN','KPI-005/027/028/030/033','Returned so returns; hai trục ngày','Không xác nhận chi hoàn/lần trả vật lý','LINK-007/008/009; SRC-014'),
('Bán/trả → kho','inventory; products; orders/items; returns','product_id–tháng thực nghiệm; snapshot duy nhất','KPI-034/035/036/039/047','Giao cặp có mặt; residual gap tháng1','Không bridge xuất/thực nhập; không panel đầy đủ/kiểm kê','LINK-009–012')]
p013_show([dict(zip(['Mắt xích khái niệm','Bảng','Khóa đã xác minh','KPI','Có thể đối chiếu','Không truy vết/kết luận được','Nguồn'],r)) for r in p013_flow],'Luồng dữ liệu phục vụ kể chuyện — chưa là chuỗi sự kiện đầy đủ')
```

## TH01.3.11 — Kết luận được hỗ trợ và câu hỏi chưa có lời giải

CL truy ngược LINK/SRC. A là số học/định nghĩa theo phạm vi đã kiểm, B là mô tả có giới hạn, C là giả thuyết cần kiểm tra, D là chưa đủ cơ sở. Trạng thái KPI theo mục đích được giữ nguyên. C/D được trình bày riêng, không đưa vào danh sách kết luận chắc chắn.

```python
p013_cls=[]
for i,x in enumerate(p013_links,1):
    level='A' if x['Kiểu liên kết']=='L1' else 'C' if x['Kiểu liên kết']=='L5' else 'B'
    p013_cls.append({'Mã kết luận':f'CL-{i:03}','Nhóm A/B/C/D':level,'Nội dung':x['Kết luận được phép'],
    'Nghiệp vụ liên quan':'; '.join(sorted({x['Nghiệp vụ A'],x['Nghiệp vụ B']})),'Bằng chứng':x['Bằng chứng định lượng'],
    'Nguồn':x['Mã liên kết']+'; '+x['Phát hiện A']+'; '+x['Phát hiện B']+'; '+x['Nguồn bổ sung'],
    'Phạm vi':x['Đơn vị và phạm vi'],'Chỉ tiêu KPI liên quan':x['Mã KPI'],'Mức độ xác minh':x['Trạng thái xác minh']+'; '+x['Mức KPI giữ nguyên'],
    'Giới hạn':x['Kết luận không được phép']+'; '+x['Vấn đề đang mở'],'Câu hỏi tiếp theo':x['Hướng kiểm tra tiếp']})
p013_cls.append({'Mã kết luận':'CL-019','Nhóm A/B/C/D':'D','Nội dung':'Chưa đủ cơ sở kết luận đã truy vết đầy đủ quảng cáo → đơn → thu tiền → giao → hoàn → thực nhập kho',
 'Nghiệp vụ liên quan':'NV01; NV02; NV03; NV04; NV05; NV06','Bằng chứng':p013_findex['SRC-010']['Bằng chứng định lượng']+'; '+p013_findex['SRC-017']['Bằng chứng định lượng']+'; '+p013_m('NV06','M69'),
 'Nguồn':'LINK-007/009/013/015; SRC-009/010/017/024; TH01.2 mục tiêu bị chặn','Phạm vi':'Nguồn hiện tại; thiếu bridge sự kiện, quyết toán, chứng từ kho',
 'Chỉ tiêu KPI liên quan':'KPI-004; KPI-005; KPI-030; KPI-038; KPI-059; KPI-068','Mức độ xác minh':'D — Thiếu bằng chứng cho mục tiêu truy vết/hiệu quả',
 'Giới hạn':'Chưa conversion/CAC/ROI/ROAS, thực thu/lợi nhuận hoặc tối ưu từ tương quan/cờ','Câu hỏi tiếp theo':'Bổ sung định nghĩa, bridge, exposure/chi phí, chứng từ và thiết kế nghiên cứu'})
p013_show([r for r in p013_cls if r['Nhóm A/B/C/D'] in ['A','B']],'CL nhóm A/B — dùng với đúng phạm vi và chú thích')
p013_show([r for r in p013_cls if r['Nhóm A/B/C/D'] in ['C','D']],'CL nhóm C/D — giả thuyết hoặc chưa đủ cơ sở, không là kết luận chắc chắn')
```

```python
p014_doc=load_notebook_for_audit(TARGET)
p014_guard=p014_doc['cells'][106]['metadata']['th014_protection']
p014_checks=[];p014_tables=[]
def p014_check(name,ok,evidence,state=None):
    state=state or ('Đạt' if bool(ok) else 'Không đạt')
    p014_checks.append({'Kiểm chứng':name,'Trạng thái':state,'Phương pháp':'Tự động' if state in ['Đạt','Không đạt'] else 'Xác nhận nghiệp vụ ngoài dữ liệu','Bằng chứng/giới hạn':evidence})
    if state=='Không đạt':raise AssertionError(name)
def p014_show(rows,title):
    p014_tables.append(title);display(Markdown('**'+title+'**'))
    with pd.option_context('display.max_rows',None,'display.max_columns',None,'display.max_colwidth',None):display(pd.DataFrame(rows))
def p014_records(pos,required):
    found=[]
    for out in p014_doc['cells'][pos-1].get('outputs',[]):
        html=out.get('data',{}).get('text/html')
        if not html:continue
        parser=Tables();parser.feed(''.join(html))
        for table in parser.tables:
            drop=table[0][0]=='';head=table[0][1:] if drop else table[0]
            if not set(required)<=set(head):continue
            found.extend(dict(zip(head,r[1:] if drop else r)) for r in table[1:])
    assert found,(pos,required)
    return found
p014_cross=p014_records(24,['Mã đối soát','Kết quả đối soát'])
p014_kparts=p014_records(36,['Mã chỉ tiêu']);p014_saved_kpis={}
for r in p014_kparts:p014_saved_kpis.setdefault(r['Mã chỉ tiêu'],{}).update(r)
p014_rules=p014_records(64,['Mã','Nhóm A/B/C/D'])
p014_decisions=p014_records(60,['Mã gốc','Trạng thái xử lý'])
p014_k4=p014_records(62,['Mã gốc','Vì sao chưa chốt'])
p014_src=p014_records(76,['Mã phát hiện nguồn','Mã gốc'])
p014_links=p014_records(90,['Mã liên kết','Kiểu liên kết'])
p014_cls=p014_records(94,['Mã kết luận','Nhóm A/B/C/D'])
p014_warnings=p014_records(96,['Trường hợp','Cách hiểu dễ nhầm'])
p014_previous_handoff=p014_records(102,['Chủ đề','Hướng đưa vào báo cáo tổng hợp'])
p014_previous_directions=p014_records(100,['Hướng đồ án','Trạng thái triển khai'])
manual_columns = ['Nội dung', 'Trạng thái', 'Kết quả/giới hạn', 'Phạm vi yêu cầu']
p014_manual0 = [dict(zip(manual_columns, row)) for row in p013_manual]
p014_saved_checks = p013_checks.copy()
p014_unconfirmed=[r for r in p014_saved_checks if r['Trạng thái']=='Không thể chạy']
p014_saved_conclusions=[''.join(o['data']['text/markdown']) for o in p014_doc['cells'][103]['outputs'] if 'text/markdown' in o.get('data',{})]
p014_ki={r['Mã chỉ tiêu']:r for r in copy.deepcopy(p012_kpis)}
p014_si={r['Mã phát hiện nguồn']:r for r in p014_src};p014_li={r['Mã liên kết']:r for r in p014_links}
p014_ci={r['Mã kết luận']:r for r in p014_cls};p014_di={r['Mã gốc']:r for r in p014_decisions}
p014_freeze={k:p012_digest(v) for k,v in {'KPI':p012_kpis,'RULE':p012_rules,'D':p012_decision_rows,'K':p012_cross,'SRC':p013_findings,'LINK':p013_links,'CL':p013_cls}.items()}
p014_fin={r['Chỉ tiêu']:r['Tính từ CSV (ĐVT)'] for r in p014_records(12,['Chỉ tiêu','Tính từ CSV (ĐVT)'])}
p014_kcounts=Counter(r['Kết quả đối soát'] for r in p014_cross)
p014_xcounts=Counter(r['Mức xác minh'] for r in p014_ki.values())
p014_dcounts=Counter(r['Trạng thái xử lý'] for r in p014_decisions)
p014_lcounts=Counter(r['Kiểu liên kết'] for r in p014_links)
p014_ccounts=Counter(r['Nhóm A/B/C/D'] for r in p014_cls)
p014_old_result=p014_doc['cells'][72]['metadata']['th013_execution']['result']
p014_old_table_count = p013_result['total_tables']
p014_themes=[{'Ô bảng nguồn':pos,'Tiêu đề chủ đề':''.join(p014_doc['cells'][pos-2]['source']).splitlines()[0]} for pos in [78,80,82,84,86,88]]
p014_check('Đầu vào đúng106 ô',p014_guard['cells']==106,'Ảnh chụp trước khi nối TH01.4, không suy từ số ô đã mở rộng.')
p014_check('dữ liệu notebook106 ô và metadata cũ bảo toàn',notebook_prefix(p014_doc['cells'][:106]) == notebook_prefix(initial_notebook['cells'][:106]) and stable_digest(p014_doc['metadata']) == stable_digest(initial_notebook['metadata']),'Gồm source/output/metadata/ID/execution_count; root metadata giữ nguyên.')
p014_inventory=[
('Sáu notebook NV01–NV06',len(NB),'Sáu nguồn duy nhất; chỉ đọc','TH01.1 ô 6; đường dẫn/hash trong nguồn'),
('Đối soát K1–K5',len(p014_cross),'; '.join(k+'='+str(p014_kcounts[k]) for k in ['K1','K2','K3','K4','K5']),'TH01.1 ô 24'),
('Danh mục KPI',len(p014_ki),'; '.join(x+'='+str(p014_xcounts[x]) for x in ['X1','X2','X3','X4']),'TH01.2 ô 36; bốn phép chiếu cùng danh mục'),
('Quy tắc RULE',len(p014_rules),'Nguyên định nghĩa, phạm vi và mức','TH01.2 ô 64'),
('Cặp không thay thế',len(p012_pairs),'Không hợp nhất các nghĩa khác nhau','TH01.2 ô 58'),
('Vấn đề V/D',len(p014_decisions),'; '.join(d+'='+str(p014_dcounts[d]) for d in ['D1','D2','D3','D4','D5']),'TH01.2 ô 60'),
('K4 còn mở',len(p014_k4),'; '.join(r['Mã gốc'] for r in p014_k4),'TH01.1 ô 24; TH01.2 ô 62'),
('Phát hiện SRC',len(p014_src),'Có mã gốc, ô, số liệu, đơn vị, phạm vi, mẫu số','TH01.3 ô 76'),
('Liên kết LINK',len(p014_links),'; '.join(k+'='+str(p014_lcounts[k]) for k in ['L1','L2','L3','L4','L5']),'TH01.3 ô 90'),
('Kết luận CL',len(p014_cls),'; '.join(k+'='+str(p014_ccounts[k]) for k in ['A','B','C','D']),'TH01.3 ô 94'),
('Cảnh báo diễn giải',len(p014_warnings),'Đơn vị, tiền, kho, kênh và giới hạn tác động','TH01.3 ô 96'),
('Nhóm chủ đề liên nghiệp vụ',len(p014_themes),'Sáu chủ đề đã có ở TH01.3.3–8; không cộng với số NV thành tổng','TH01.3 ô 78/80/82/84/86/88'),
('Nhóm bàn giao / hướng đồ án',str(len(p014_previous_handoff))+' / '+str(len(p014_previous_directions)),'Bàn giao có điều kiện, chưa thực thi giai đoạn sau','TH01.3 ô 100/102'),
('Đoạn kết luận liên nghiệp vụ',len(p014_saved_conclusions),'Đã đọc; TH01.4 tổng kết ngắn gọn theo mục đích báo cáo','TH01.3 ô 104'),
('Xác nhận nghiệp vụ chưa chạy',len(p014_unconfirmed),'Không thể chạy do thiếu định nghĩa/chứng từ; không tính đạt','TH01.3 ô 106; cùng bốn K4'),
('Nhóm rà soát thủ công',len(p014_manual0),'Theo dõi riêng; không chứng minh nghiệp vụ ngoài CSV','TH01.3 ô 106'),
('Kiểm chứng tự động trước TH01.4',p014_old_result['total_passed'],'Kết quả lịch sử được đối chiếu tái thực thi; không cộng lượt chạy','Metadata ô 73; kiểm chứng ô 30/72/106')]
p014_show([dict(zip(['Đầu vào','Số lượng đọc được','Trạng thái/phạm vi','Nguồn truy ngược'],r)) for r in p014_inventory],'Kiểm kê đầu vào TH01.4 — số đọc từ kết quả hiện có, mốc chỉ dùng đối chiếu')
```

## TH01.4.6 — Những giới hạn quan trọng nhất

15 hàng dưới là giới hạn **theo cách sử dụng**, không là 15 lỗi dữ liệu độc lập. ## TH01.4.10 — Điều kiện bàn giao sang Giai đoạn 2

**A:** quy ước phân tích có thể kế thừa trong phạm vi đã kiểm; **B:** chính sách/quyết định cần thống nhất trước nhiệm vụ xử lý tương ứng; **C:** thông tin cần bổ sung. Không coi các đề xuất B/C là chính sách đã được nhóm phê duyệt. Giai đoạn 2.1 dự kiến làm sạch/xử lý và 2.2 dự kiến xây dựng đặc trưng/bảng, nhưng TH01.4 chỉ lập điều kiện, chưa thực hiện thao tác nào.

Các mã phụ thuộc cho biết công việc cần trước **một mục tiêu cụ thể**, không chặn toàn bộ Giai đoạn 2. Có thể tiếp tục mô tả Net khi chưa có quyết toán; không dùng Net làm thực thu. Khóa và công thức vẫn phải được kiểm lại khi nguồn cập nhật. Tiêu chí hoàn thành là bằng chứng/biên bản cần có ở bước sau, không phải trạng thái đã đạt tại TH01.4.

## TH01.4.14 — Kết quả cuối cùng của TH01

### TH01.5.1 — Kết luận sáu nghiệp vụ và bảy insight trọng tâm

Câu hỏi, số liệu, công thức, ý nghĩa và giới hạn được trình bày cùng nguồn. Nhận định mới không nâng quan hệ quan sát thành quan hệ nhân quả; phân phối biên không chứng minh giao nhóm khách–sản phẩm.


**NV01 — kết luận hiện hành [I01, I02]**

Revenue/COGS theo lịch và Net sản phẩm trả lời hai câu hỏi khác nhau. Theo dõi đồng thời xu hướng lịch sử, khoảng cách Revenue−COGS và cơ cấu lượng/Net delivered; không gọi khoảng cách này là lợi nhuận ròng. Mùa vụ chỉ gợi ý lịch rà soát, chưa là dự báo hay lệnh nhập hàng.

**NV02 — kết luận hiện hành [I03]**

Tần suất và giá trị đơn cần theo dõi riêng trên cùng cửa sổ. Nhóm nhiều đơn đóng góp đáng kể Net nhưng AOV gộp không cao nhất. Nhóm tuổi 25–34 có 36.342 khách (29,81%); tỷ lệ có đơn theo tuổi chỉ 73,75–74,39%. Kênh organic_search có 36.450 hồ sơ (29,89%), không chứng minh hiệu quả thu hút. Không chuyển các nhóm mô tả thành cụm RFM hay quyết định ngân sách.

**NV03 — kết luận hiện hành [I04]**

Delivered chiếm đa số nhãn đơn, còn Payment=Net là đối chiếu giá trị ghi nhận. Cần xác nhận ngữ nghĩa trạng thái, mốc giao và phạm vi shipment trước khi đo chất lượng vận hành hoặc SLA; chênh COD chỉ dùng để chọn mẫu kiểm tra.

**NV04 — kết luận hiện hành [I05]**

Ưu tiên xác minh nhóm Refund vượt Payment và nhãn wrong_size theo hai luồng tài chính/chất lượng dữ liệu. Không đổi số dòng returns thành số đơn, không gán nguyên nhân gốc hoặc kết luận mất tiền khi chưa có chứng từ.

**NV05 — kết luận hiện hành [I06]**

Độ phủ và tập sản phẩm thay đổi là điều kiện đọc biến động snapshot. Theo dõi phần tập chung cùng thành phần mẫu; residual bằng 0 và tương quan cao chưa đủ thay units_sold bằng quantity hay xác nhận kho vật lý.

**NV06 — kết luận hiện hành [I07]**

Nhóm có mã có mean quantity nhỉnh hơn, cùng median 5, nhưng mean Net thấp hơn. Hiện diện mã, Discount, Payment và hiệu quả marketing là các khái niệm riêng; chưa có căn cứ mở rộng khuyến mãi hoặc đánh giá ROI từ so sánh quan sát này.

### I01 — Suy giảm bán hàng lan rộng theo thời gian; cần kiểm tra cả COGS

**Câu hỏi:** Sự giảm quy mô có chỉ nằm ở một vài kỳ, và Revenue có phản ánh đầy đủ chênh lệch với COGS không?

**Bằng chứng:** Revenue toàn kỳ 16.430.476.585,53 ĐVT, COGS 14.163.450.519,17 ĐVT. Năm 2019 Revenue giảm 38,56% và cả 12/12 tháng giảm YoY. Năm 2021 Revenue giảm 1,09% trong khi COGS tăng 6,21%; tỷ lệ (Revenue−COGS)/Revenue còn 9,77%. Tháng 04–06 có mức Revenue/ngày tương đối trên nền năm trong 10/10 năm đầy đủ 2013–2022.

**Nguồn:** [NV01 ô 90 (45a13b3c)](NV01_phan_tich_ban_hang.ipynb#45a13b3c) — Tableau KPI NV01.5; 12 chỉ tiêu lịch sử; [NV01 ô 92 (54489d4e)](NV01_phan_tich_ban_hang.ipynb#54489d4e) — F01–F07; nguồn bảng/biểu đồ NV01.1–4

**Phạm vi/công thức:** sales theo ngày, mọi trạng thái trong phạm vi đối soát Gross; 04/07/2012–31/12/2022, chỉ so YoY các kỳ đủ lịch. Chỉ số mùa vụ = 100×(Revenue tháng/số ngày tháng)/(Revenue năm/số ngày năm). COGS là trường nguồn sales; không thay bằng giá vốn danh mục.

**Ý nghĩa kinh doanh:** Dùng bảng theo tháng và khoảng cách với COGS làm điểm khởi đầu rà soát giá, lượng, cơ cấu và phạm vi ghi nhận; không chỉ chọn một tháng xấu để giải thích cả năm.

**Mức chắc chắn:** Cao về số học và mô tả lịch sử; nguyên nhân suy giảm chưa xác minh.

**Giới hạn/giả thuyết:** 2012 thiếu lịch; giá trị danh nghĩa trước giảm giá, chưa quyết toán và thiếu chi phí khác. Mùa vụ lịch sử không bảo đảm nhu cầu tương lai; không gọi Revenue−COGS là lợi nhuận ròng.

**Truy vết đóng góp:** Kế thừa NV01.1–5 của nhóm; Bước 4 bổ sung thứ tự câu hỏi điều hành.

### I02 — Danh mục ưu tiên thay đổi theo lượng và Net

**Câu hỏi:** Nên theo dõi mã và nhóm sản phẩm nào khi mục tiêu là lượng hoặc giá trị Net?

**Bằng chứng:** Delivered có 2.565.930 đơn vị, Net 12.518.175.957,20 ĐVT. Mã 791 đứng đầu lượng (36.517), mã 487 đứng đầu Net (303.232.685,02 ĐVT). Streetwear chiếm 55,084628% lượng và 80,123890% Net; Outdoor tương ứng 36,391679% và 14,984842%. Activewear chiếm 32,274107% lượng nhưng 12,302408% Net; Everyday và Balanced đóng góp lần lượt 32,912655% và 31,175678% Net. 296 mã đạt 80,000117% lượng, bằng 18,793651% của 1.575 mã có delivered. Riêng top 10 theo Net đóng góp 1,951,688,030.71 ĐVT, bằng 15.59% Net delivered; đây là tỷ trọng giá trị được tính bổ sung ở Bước 4, khác ngưỡng 80% lượng.

**Nguồn:** [NV01 ô 100 (nv01-step3-02)](NV01_phan_tich_ban_hang.ipynb#nv01-step3-02) — NV01.6: độ phủ và hai phạm vi tiền/lượng; [NV01 ô 102 (nv01-step3-04)](NV01_phan_tich_ban_hang.ipynb#nv01-step3-04) — NV01.6: hai bảng top10 và biểu đồ lượng/Net; [NV01 ô 104 (nv01-step3-06)](NV01_phan_tich_ban_hang.ipynb#nv01-step3-06) — NV01.6: category/segment và biểu đồ tỷ trọng lượng/Net; [NV01 ô 106 (nv01-step3-08)](NV01_phan_tich_ban_hang.ipynb#nv01-step3-08) — NV01.6: ngưỡng 80% quantity và đường tích lũy

**Phạm vi/công thức:** Dòng hàng GD2 lọc order_status=delivered theo ngày đặt toàn kỳ; tổng hợp product_id, category/segment hiện tại. Net dòng = quantity×unit_price−discount_amount. Tỷ trọng Net = tổng Net nhóm/tổng Net delivered; top 10 được chọn theo Net giảm dần. Ngưỡng 80% tính trên quantity, giữ mã vượt ngưỡng; không JOIN thêm bảng kho.

**Ý nghĩa kinh doanh:** Lập hai danh sách theo dõi sản phẩm phục vụ lượng và giá trị; danh sách Net tập trung vào nhóm khác với danh sách lượng. Chưa đủ căn cứ tăng tồn hoặc loại bỏ mã chưa ghi nhận bán.

**Mức chắc chắn:** Cao về tổng hợp delivered; mức ưu tiên thương mại còn phụ thuộc biên đóng góp và khả dụng.

**Giới hạn/giả thuyết:** Catalogue hiện tại không có hiệu lực lịch sử; 814 mã không có dòng đặt và 23 mã có đặt nhưng chưa delivered không đồng nghĩa kém hiệu quả. Không suy tập mã bán trùng tập có inventory, không đánh giá lợi nhuận từ Net.

**Truy vết đóng góp:** Phương pháp xếp hạng lượng, category, ngưỡng 80% kế thừa Thu Luyn EDA ô 57–75, đã hiệu chỉnh ở NV01.6 Bước 3; Net/segment do nhóm bổ sung Bước 3. Tỷ trọng top 10 Net và diễn giải ưu tiên là Bước 4.

### I03 — Giá trị tích lũy cao gắn với nhiều đơn hơn, chưa phải AOV cao hơn

**Câu hỏi:** Có nên dùng tần suất tích lũy để xác định nhóm cần chăm sóc hoặc định giá?

**Bằng chứng:** Trong 90.246 khách có đơn, nhóm ≥21 đơn có 6.883 khách (7,63%) và 32,10% Net, AOV gộp 24.119,97 ĐVT; nhóm 6–20 đơn có AOV 24.496,40 ĐVT. Toàn bộ đơn có AOV Net gộp 24.238,33 ĐVT; delivered 24.226,41 ĐVT trên 85.115 khách/516.716 đơn. Spearman số đơn–Net 0,9132, số đơn–AOV 0,1753. Organic_search có 36.450 hồ sơ (29,89% customers), tỷ lệ có đơn giữa các kênh chỉ 73,63–74,84%.

**Nguồn:** [NV02 ô 55 (a49e5e29)](NV02_phan_tich_khach_hang.ipynb#a49e5e29) — NV02.3: nhóm tần suất, AOV và quan hệ hạng; [NV02 ô 57 (07491000)](NV02_phan_tich_khach_hang.ipynb#07491000) — NV02.3: độ nhạy A/B và tổng Net delivered; [NV02 ô 73 (b82f5d38)](NV02_phan_tich_khach_hang.ipynb#b82f5d38) — NV02.4: bảng và biểu đồ acquisition_channel; [NV02 ô 84 (22c663dd)](NV02_phan_tich_khach_hang.ipynb#22c663dd) — NV02.5: KPI khách, nhóm tuổi/kênh và giới hạn thời gian

**Phạm vi/công thức:** A: toàn bộ 646.945 đơn, 90.246 khách có đơn; B: delivered, mẫu khách riêng. AOV gộp=ΣNet/Σđơn; mean AOV/khách=mean(Net_i/n_i), không thay thế nhau. Nhóm tần suất mô tả rời nhau trên toàn lịch sử; acquisition_channel là thuộc tính hồ sơ, không là nguồn tạo từng đơn.

**Ý nghĩa kinh doanh:** Chăm sóc khách cần nhìn tần suất, AOV và thời gian quan sát riêng; không mặc định tăng giá hoặc ưu đãi lớn cho nhóm có nhiều đơn, không phân bổ ngân sách chỉ theo tổng Net kênh.

**Mức chắc chắn:** Cao về mô tả khách và định nghĩa AOV; giả thuyết về phản ứng chăm sóc chưa được thử nghiệm.

**Giới hạn/giả thuyết:** Net_i=n_i×AOV_i tạo liên hệ cơ học. Median Recency A là 991 ngày tại 31/12/2022; 477.453 đơn trước signup chưa có giải thích nghiệp vụ. Không coi signup là bắt đầu hành trình, Recency cao là rời bỏ, hoặc nhóm mô tả là kết quả phân cụm.

**Truy vết đóng góp:** Kế thừa NV02.1–5 và đối chiếu thời gian NV06.5; Bước 4 bổ sung cách chọn phép đo chăm sóc.

### I04 — Chênh trạng thái theo thanh toán là tín hiệu chọn mẫu kiểm tra

**Câu hỏi:** Nhóm nào cần kiểm tra hồ sơ vận hành trước khi đổi chính sách giao hoặc thanh toán?

**Bằng chứng:** Delivered 516.716/646.945 đơn (79,87%). Payment=Net khớp trên toàn bộ đơn nhưng chưa xác nhận thực thu. Có 566.067 đơn có shipment và 80.878 đơn không có; 80.314 đơn thiếu thuộc cancelled/created/paid. Với COD so non-COD, sau cùng trọng số năm 2013–2022, cancelled cao hơn 8,050 điểm phần trăm và returned cao hơn 3,894 điểm phần trăm.

**Nguồn:** [NV03 ô 77 (nv034-12)](NV03_phan_tich_don_hang.ipynb#nv034-12) — NV03.4: COD/non-COD với cùng trọng số năm; [NV03 ô 87 (nv035-02)](NV03_phan_tich_don_hang.ipynb#nv035-02) — NV03.5: chỉ tiêu đơn, payment và shipment; [NV03 ô 89 (nv035-04)](NV03_phan_tich_don_hang.ipynb#nv035-04) — NV03.5: F01–F09 cùng chứng cứ/giới hạn

**Phạm vi/công thức:** Cấp order_id, trạng thái snapshot; tỷ lệ trạng thái/N đơn của từng phương thức, chuẩn hóa cùng trọng số năm trên 2013–2022. Shipment chỉ đo các đơn hiện diện; chưa đồng nghĩa ngày sự kiện thực tế.

**Ý nghĩa kinh doanh:** Chọn mẫu COD và non-COD cùng kỳ để kiểm tra trạng thái, chứng từ và lý do; theo dõi độ phủ shipment theo nhãn trước khi đo SLA hoặc hạn chế phương thức thanh toán.

**Mức chắc chắn:** Cao về đối chiếu ghi nhận và chênh quan sát; không xác nhận phương thức thanh toán gây hủy/trả.

**Giới hạn/giả thuyết:** Chưa kiểm soát sản phẩm, khách, vùng hay mức giá. Thiếu shipment không chứng minh chưa giao; cuối kỳ có thể chưa đủ vòng đời. Payment hiện diện ở đơn cancelled không tự là tiền thực thu.

**Truy vết đóng góp:** Kế thừa NV03.1–5 và đối soát TH01; Bước 4 nối tín hiệu kiểm tra với điều kiện hành động.

### I05 — Tách kiểm tra hoàn tiền và xác minh nhãn lý do hoàn trả

**Câu hỏi:** Đâu là nhóm cần ưu tiên rà chứng từ và đâu là giả thuyết cần xác minh trước can thiệp?

**Bằng chứng:** 39.939 dòng returns thuộc 36.062 đơn; wrong_size có 13.967 dòng (34,97% dòng returns). Có 3.640 đơn Refund cao hơn Payment quá dung sai 0,05 ĐVT, tổng phần vượt trong nhóm này 4.308.810,51 ĐVT. Có 80 đơn nhãn returned chưa có dòng returns, ngày đặt 06–31/12/2022.

**Nguồn:** [NV04 ô 30 (nv042-06)](NV04_phan_tich_hoan_tra_hoan_tien.ipynb#nv042-06) — NV04.2: nhóm Refund/Payment cấp đơn; [NV04 ô 105 (nv045-02)](NV04_phan_tich_hoan_tra_hoan_tien.ipynb#nv045-02) — NV04.5: chỉ tiêu hoàn trả/hoàn tiền và wrong_size; [NV04 ô 107 (nv045-04)](NV04_phan_tich_hoan_tra_hoan_tien.ipynb#nv045-04) — NV04.5: F01–F10; 80 returned thiếu đối ứng

**Phạm vi/công thức:** Sổ cấp order_id: cộng Refund quan sát từ returns trước khi đối chiếu Payment; chỉ nhóm có returns, mọi nhãn đơn theo nguồn. Wrong_size tính trên dòng returns, không chia tổng đơn. Ngày trả là trục sự kiện riêng với ngày đặt; không trừ Refund toàn kỳ khỏi Net delivered để gọi thực thu.

**Ý nghĩa kinh doanh:** Ưu tiên danh sách hoàn tiền vượt để rà chứng từ; đồng thời lấy mẫu wrong_size để kiểm tra nhãn, hướng dẫn kích cỡ và đối ứng dòng bán. Hai luồng này chưa chứng minh chung một nguyên nhân.

**Mức chắc chắn:** Cao về chênh số học và tần suất nhãn; nguyên nhân, số tiền thực hoàn và tổn thất chưa xác minh.

**Giới hạn/giả thuyết:** Một đơn có thể nhiều dòng/lý do; không nhân Payment theo dòng trả. Mức vượt không là thất thoát và wrong_size chưa chứng minh lỗi bảng size hay lỗi sản phẩm; thiếu cuối kỳ chưa là mất dữ liệu.

**Truy vết đóng góp:** Kế thừa NV04.1–5, liên hệ ledger NV03/NV01; Bước 4 tách hai luồng xác minh/hành động.

### I06 — Ưu tiên độ phủ và nghĩa trường trước quyết định tồn kho

**Câu hỏi:** Biến động tổng kho và chênh lượng bán có đủ làm căn cứ bổ sung tồn hoặc kết luận thất thoát?

**Bằng chứng:** Inventory có 60.247 dòng, 1.624/2.412 sản phẩm và chỉ 22 mã đủ 126 tháng; 25.456 vị trí tháng vắng giữa kỳ đầu–cuối. Ví dụ 2020-12→2021-01: chênh tổng −12.473, phần tập chung +246, phần thành phần mẫu −12.719. Trong 56.615 cặp sản phẩm–tháng giao với quantity mọi trạng thái, chỉ 1.716 cặp bằng nhau (3,030999%) dù Pearson 0,989704. 51.524/51.524 cặp snapshot liền tháng có residual 0; 7.099 khoảng qua tháng vắng không nối.

**Nguồn:** [NV05 ô 121 (nv055-c4374b05)](NV05_phan_tich_hang_ton_kho.ipynb#nv055-c4374b05) — NV05.5: M01–M60, độ phủ/phân rã/đối chiếu lượng/residual

**Phạm vi/công thức:** Inventory cấp product_id×tháng snapshot 07/2012–12/2022. Quantity order_items gộp cùng product_id×tháng đặt, mọi trạng thái trong phép so 56.615 cặp; không là lượng giao/nhập kho. Snapshot là mức, received/sold là lượng nguồn; residual chỉ tính cặp liền tháng.

**Ý nghĩa kinh doanh:** Cần bảng độ phủ và phân rã thành phần mẫu khi đọc biến động; chỉ dùng danh sách sản phẩm NV01 để đặt câu hỏi rà soát, chưa phát lệnh nhập hay cắt tồn từ hai nguồn này.

**Mức chắc chắn:** Cao về khóa, độ phủ và phép so trong phạm vi; nghiệp vụ kho vật lý chưa xác minh.

**Giới hạn/giả thuyết:** Không có cấp kho/địa điểm, thời điểm kiểm kê thực và quy tắc bán. Thiếu snapshot không là hết hàng; tương quan cao/residual 0 không xác nhận kho thật; units_sold không thay quantity, chênh không là thất thoát.

**Truy vết đóng góp:** Kế thừa NV05.1–5; Bước 4 bổ sung điều kiện nối với cơ cấu sản phẩm NV01.6.

**Ý nghĩa kinh doanh:** Chưa mở rộng mã chỉ vì lượng bình quân cao hơn; cần xem đồng thời Net và mẫu số, sau đó thiết kế thử nghiệm có nhóm đối chứng và dữ liệu chi phí. Quan hệ web nhạy với nền lịch, cần đo attribution riêng.

**Truy vết đóng góp:** Mean/median quantity kế thừa Thu Luyn EDA ô 84/86/88, kiểm chứng trong NV06.6 Bước 3. Net/status/web kế thừa nhóm; Bước 4 kết nối điều kiện đánh giá hiệu quả.

### TH01.5.3 — Hành động, KPI đánh giá và thứ tự ưu tiên

P0 ưu tiên chứng từ/định nghĩa; P1 gồm theo dõi có thể dùng ngay và kiểm tra giả thuyết; P2 chỉ triển khai thử khi đủ điều kiện. Đây là mức ưu tiên đề xuất, chưa là quyết định của doanh nghiệp. KPI thử nghiệm có mẫu số mới, không thay KPI mô tả lịch sử. Chưa ấn định mức tăng doanh thu/giảm chi phí vì không có ước lượng hợp lệ.


### KN01 — Lập hàng đợi rà chứng từ hoàn tiền (P0; Cần xác minh)

**Vấn đề/bằng chứng:** 3.640 đơn có Refund vượt Payment, tổng phần vượt 4.308.810,51 ĐVT; ý nghĩa Payment/Refund chưa được xác nhận. [I04, I05]

**Hành động:** Đối soát danh sách cấp đơn với giao dịch thu, hoàn, điều chỉnh, phí/thuế và lịch sử trạng thái; ghi mã chứng từ, nguyên nhân xác minh, người phụ trách và trạng thái xử lý từng đơn. Không sửa số tiền nguồn hoặc tự giảm Refund.

**Kỳ vọng:** Có phân loại chênh có chứng từ và nhóm chưa giải thích; chưa ước lượng thu hồi tiền hay giảm tổn thất.

**KPI theo dõi:** Tỷ lệ hồ sơ đã rà=đơn có kết luận kèm chứng từ/3.640; tổng phần vượt chưa giải thích=Σmax(Refund−Payment,0) của hồ sơ chưa giải thích, theo cùng dung sai 0,05 ĐVT.

**Điều kiện/cách đánh giá:** Phải có giao dịch settlement, mã hoàn tiền và chính sách khoản điều chỉnh; kiểm tra cả hồ sơ đơn lẫn nhiều dòng returns, không đánh đồng hoàn tất rà soát với hợp lệ tài chính.

**Lý do ưu tiên:** Ngăn diễn giải sai dòng tiền và xử lý nhóm chênh đã định lượng; không đòi suy đoán nguyên nhân.

### KN02 — Chuẩn hóa định nghĩa và độ phủ vận hành (P0; Cần xác minh)

**Vấn đề/bằng chứng:** Trạng thái, shipment, snapshot, units_sold và dữ liệu web chưa đủ chứng từ để làm chính sách/SLA. [I04, I06, I07]

**Hành động:** Lập hồ sơ định nghĩa có chủ sở hữu: mốc actual/planned, timestamp chốt trạng thái, lịch snapshot và phạm vi kho, nguồn units_sold, thang bounce_rate/attribution. Theo dõi riêng DS027/DS030/DS037/DS038; chỉ đóng khi có tài liệu và phép đối chiếu phù hợp.

**Kỳ vọng:** Tách được thiếu bao phủ, khác định nghĩa và lỗi thực sự; chưa tuyên bố giảm thời gian giao hoặc tồn kho.

**KPI theo dõi:** Tỷ lệ trường trọng yếu có định nghĩa được duyệt=trường có tài liệu và chủ sở hữu/tập trường rà soát đã chốt; tỷ lệ shipment hiện diện theo từng status=N đơn có shipment/N đơn status; độ phủ snapshot=số cặp product×tháng có mặt/số cặp phải có theo lịch nghiệp vụ đã duyệt.

**Điều kiện/cách đánh giá:** Chốt mẫu số phải có trước khi gọi thiếu dữ liệu; không dùng tích toàn catalogue×toàn lịch làm kỳ vọng mặc định; bốn hồ sơ vẫn mở ở thời điểm Bước 4.

**Lý do ưu tiên:** Là điều kiện đo đúng hiệu quả các hành động tài chính, kho và marketing tiếp theo.

### KN03 — Theo dõi bán hàng bằng hai trục lượng và giá trị (P1; Trực tiếp)

**Vấn đề/bằng chứng:** Xếp hạng lượng và Net khác nhau; biến động Revenue và COGS có thể trái hướng, độ phủ kho không đồng nhất. [I01, I02, I06]

**Hành động:** Dùng bảng/biểu đồ NV01 hiện có trong cuộc rà soát định kỳ: theo dõi Revenue/COGS và Net delivered riêng, hai top sản phẩm, tỷ trọng category/segment; kèm mẫu số và độ phủ. Yêu cầu người phụ trách xác nhận giá, khả dụng và vòng đời trước mọi quyết định nhập/cắt mã.

**Kỳ vọng:** Thống nhất cách đọc quy mô, cơ cấu và danh sách kiểm tra; chưa cam kết tăng doanh thu hoặc tối ưu tồn.

**KPI theo dõi:** Theo kỳ đủ lịch: YoY Revenue; (Revenue−COGS)/Revenue theo sales; tỷ trọng Net delivered nhóm=ΣNet nhóm/ΣNet delivered; lượng delivered và số mã có delivered; top10 Net share tách ngưỡng80% quantity.

**Điều kiện/cách đánh giá:** Cố định kỳ, trạng thái và phiên bản catalogue; dùng nhóm hiện tại cho mô tả, cần lịch sử hiệu lực cho phân rã xu hướng. Nếu chuyển sang quyết định kho phải hoàn thành KN02.

**Lý do ưu tiên:** Triển khai ngay cách theo dõi bằng số đã kiểm; rủi ro thấp, không thay công thức hay dữ liệu.

### KN04 — Thử nghiệm chăm sóc trên cửa sổ khách tương đương (P2; Cần thử nghiệm)

**Vấn đề/bằng chứng:** Nhiều đơn gắn tổng Net cao nhưng AOV không cao nhất; lịch sử tích lũy và thời gian quan sát khác nhau. [I03]

**Hành động:** Sau khi xác minh thời điểm đủ điều kiện và sự đồng ý liên hệ, chọn khách theo lịch sử đặt đã quan sát trong cửa sổ bằng nhau, tách delivered; phân nhóm ngẫu nhiên nhận/không nhận nội dung chăm sóc, giữ chính sách đo chung. Chưa tạo phân cụm RFM.

**Kỳ vọng:** Ước lượng hiệu quả chăm sóc nếu có, tránh chọn nhóm chỉ vì Net tích lũy cao; chưa dự báo mức tăng.

**KPI theo dõi:** Tỷ lệ có đơn delivered mới=N khách có ≥1 đơn delivered mới/N khách được phân nhóm; Net delivered/khách được phân nhóm; AOV=ΣNet delivered/N đơn delivered; chênh nhóm thử−đối chứng kèm khoảng tin cậy.

**Điều kiện/cách đánh giá:** Chốt cửa sổ theo dõi đủ vòng đời, cỡ mẫu, đơn vị ngẫu nhiên và tiêu chí dừng trước thử; kiểm tra cân bằng lịch sử, xử lý signup chưa rõ; không gọi nhóm khách hiện tại là churn/trung thành.

**Lý do ưu tiên:** Có ý nghĩa nhưng cần thiết kế và dữ liệu liên hệ/thời gian tin cậy; làm sau điều kiện P0.

### KN05 — Kiểm tra giả thuyết wrong_size trước thử hướng dẫn kích cỡ (P1; Cần thử nghiệm)

**Vấn đề/bằng chứng:** Wrong_size chiếm 13.967 dòng returns, nhưng nhãn chưa xác nhận nguyên nhân và chưa có mẫu số phơi nhiễm phù hợp. [I05]

**Hành động:** Lấy mẫu có phân tầng sản phẩm/kỳ để xác minh nhãn và đối ứng dòng bán; nếu hồ sơ ủng hộ giả thuyết hướng dẫn chưa phù hợp, thử nội dung kích cỡ trên tập đủ điều kiện với nhóm đối chứng. Không tự đổi mô tả nhãn hoặc trách nhiệm.

**Kỳ vọng:** Xác định khả năng cải thiện thông tin; chưa cam kết giảm tỷ lệ hoàn hay tiền hoàn.

**KPI theo dõi:** Bước xác minh: hồ sơ xác nhận nguyên nhân/tổng hồ sơ mẫu; bước thử: N đơn delivered có returns wrong_size trong cửa sổ theo dõi/N đơn delivered đủ điều kiện; theo dõi Net và tỷ lệ returns chung làm chỉ tiêu bảo vệ.

**Điều kiện/cách đánh giá:** Mẫu số thử mới khác 34,97% dòng returns lịch sử, không so trực tiếp. Chốt đối ứng đơn/dòng, cửa sổ trả đủ dài, quy tắc trạng thái và cân bằng nhóm trước thử.

**Lý do ưu tiên:** Nhãn phổ biến đủ để ưu tiên kiểm tra giả thuyết, chưa đủ để áp giải pháp đại trà.

### KN06 — Đo tác động khuyến mãi bằng thử nghiệm có đối chứng (P2; Cần thử nghiệm)

**Vấn đề/bằng chứng:** Mean quantity cao hơn nhẹ, median không đổi, mean Net thấp hơn ở nhóm mã; chưa biết phơi nhiễm và chi phí. [I03, I07]

**Hành động:** Chốt đối tượng đủ điều kiện, giá trị min_order_value, stacking và log phơi nhiễm/áp dụng; ngẫu nhiên theo khách đủ điều kiện vào nhóm có/không đề nghị mã trong cùng kỳ. Lưu assignment trước mua và phân tích theo assignment, không theo việc tự chọn dùng mã.

**Kỳ vọng:** Ước lượng Net/quantity tăng thêm có điều kiện; chỉ cân nhắc mở rộng sau kiểm tra chi phí và chỉ tiêu bảo vệ, không suy ROI hiện tại.

**KPI theo dõi:** Quantity delivered/khách được phân nhóm và Net delivered/khách được phân nhóm (kể cả khách không mua); AOV, tỷ lệ có đơn delivered, tỷ lệ returns; chênh thử−đối chứng cùng khoảng tin cậy. Chỉ tính đóng góp sau khi có COGS/chi phí phù hợp, không gọi là lợi nhuận ròng.

**Điều kiện/cách đánh giá:** Cần nhận diện phơi nhiễm, dữ liệu chi phí, ngẫu nhiên không nhiễm chéo, cỡ mẫu và kỳ theo dõi đủ vòng đời; chốt KPI chính/tiêu chí quyết định trước thử. Không dùng chênh 5,008798−4,940392 làm uplift.

**Lý do ưu tiên:** Nguy cơ hiểu sai so sánh nhóm cao; cần hoàn thành định nghĩa và thiết kế trước khi chi thêm.

### KN07 — Rà hồ sơ COD cùng kỳ trước đổi chính sách (P1; Cần xác minh)

**Vấn đề/bằng chứng:** COD có chênh cancelled/returned còn rõ sau chuẩn hóa theo năm, chưa kiểm soát khách/sản phẩm/vùng. [I04]

**Hành động:** Lấy mẫu COD và non-COD cùng kỳ, phân tầng vùng/giá trị/sản phẩm khi có đủ dữ liệu; rà lý do hủy/trả và chứng từ mốc. Chỉ đề xuất thử thay quy trình xác nhận sau khi xác minh nguyên nhân, chưa hạn chế COD đại trà.

**Kỳ vọng:** Tách tín hiệu cơ cấu khỏi nguyên nhân có thể can thiệp; chưa ước lượng số đơn giữ lại.

**KPI theo dõi:** Tỷ lệ hồ sơ có lý do xác minh=N hồ sơ có chứng từ/N hồ sơ mẫu; cancelled và returned/N đơn cùng phương thức, cùng cửa sổ đã đủ vòng đời; chênh cùng trọng số tầng, giữ N từng tầng.

**Điều kiện/cách đánh giá:** Cần lịch sử trạng thái, lý do và ngày chốt. Nếu thử quy trình mới, phân nhóm trước và theo dõi Net/tỷ lệ hoàn tất cùng chi phí xác nhận; phép chuẩn hóa năm hiện tại chưa loại trừ nhiễu khác.

**Lý do ưu tiên:** Có tín hiệu định lượng để chọn mẫu, chi phí/rủi ro chính sách chưa được ước lượng.


### TH01.5.4 — Kết luận điều hành và phạm vi sẵn sàng

Tổng kết này là phần hiện hành để dùng trong báo cáo; các phần TH01.1–4 tiếp tục giữ lịch sử đối soát và bàn giao. Đề xuất cần thử nghiệm chưa được kiểm chứng hiệu quả.


Bài toán 1 đã có nền đối soát cùng định nghĩa và dữ liệu tiền xử lý dùng chung; Bước 4 bổ sung tổng kết hiện hành từ kết quả Bước 3. Revenue trước giảm giá, Net dòng/đơn, Payment và Refund vẫn giữ nghĩa riêng. 38 đối soát và 68 KPI hiện hữu tiếp tục là căn cứ; phần tổng kết này không thay các danh mục đó.

Danh mục cần đọc theo hai mục tiêu lượng và giá trị; nhóm khách nhiều đơn cần tách giá trị tích lũy khỏi AOV. Các khác biệt theo COD và hiện diện mã là tín hiệu chọn mẫu/thiết kế thử nghiệm, chưa là tác động của chính sách. Mùa vụ lịch sử giúp đặt lịch rà soát, chưa chứng minh nhu cầu tương lai.

Ưu tiên P0 là hồ sơ hoàn tiền và định nghĩa/độ phủ; P1 gồm cách theo dõi bán hàng có thể áp dụng trực tiếp và xác minh COD/wrong_size trước can thiệp; P2 là thử nghiệm chăm sóc/khuyến mãi khi đủ điều kiện. Không có ước lượng tăng doanh thu, tiết kiệm chi phí hoặc lợi nhuận kỳ vọng đã được chứng minh.

Đủ cơ sở đưa phần phân tích mô tả, đối soát và khuyến nghị có điều kiện vào báo cáo đồ án sau kiểm tra biên tập cuối. Chưa đủ căn cứ trình bày là quyết toán tài chính, chính sách kho/SLA hay hiệu quả marketing đã xác nhận. DS027, DS030, DS037, DS038 và sáu nhóm rà soát thủ công giữ nguyên; Bài toán 2/3 không được triển khai ở bước này.
